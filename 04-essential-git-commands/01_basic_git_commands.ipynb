## 1. Git Init

### What it does
`git init` creates a new, empty Git repository in the current folder by creating a hidden `.git` directory. It does not automatically add or commit files.

### Syntax
```bash
git init
git init <directory-name>
```

### Practical example
Open a terminal in a new practice folder and run:

````python
mkdir basic-git-demo
cd basic-git-demo
git init
````

**Expected output (wording may vary by Git version):** Git reports that an empty repository was initialized.

Verify the repository:

````python
git status
````

**Expected result:** `On branch ...` and `No commits yet`; Git may also show untracked files if any exist.

**Common issue:** `git init` run in the wrong directory. Check the current path (`pwd` in Git Bash, `Get-Location` in PowerShell) before initializing.

## 2. Git Status

### What it does
`git status` shows the current branch and the state of files: untracked, modified, staged, or clean. Use it frequently before staging and committing.

### Syntax and practice


````python
git status
````

**Expected result:** Git displays the branch name and a summary such as `nothing to commit, working tree clean`, or lists changed/untracked files.

**Tip:** `git status --short` gives a compact view. `??` means untracked; ` M` means a tracked file is modified but not staged; `M ` means a modification is staged.

## 3. Git Add

### What it does
`git add` stages selected changes for the next commit. Staging lets you choose exactly what will be included.

### Syntax
```bash
git add <file>
git add .
```

### Practical example
Create a file, then stage it:

````python
# PowerShell example
Set-Content -Path notes.txt -Value 'Learning Git add'
git status
git add notes.txt
git status
````

**Expected result:** Before `git add`, `notes.txt` appears as untracked. After staging, it appears under `Changes to be committed`.

**Important:** `git add .` stages all changes under the current directory. Review `git status` first so you do not stage unwanted files.

## 4. Git Commit

### What it does
`git commit` records the staged snapshot in the repository history. A useful message briefly explains the change.

### Syntax and practice


````python
git commit -m "Add initial practice notes"
````

**Expected result:** Git creates a commit and displays its abbreviated hash, branch, and file-change summary. Exact hash and counts depend on your repository.

**Common issue:** `Author identity unknown`. Configure your identity (use your own details):
```bash
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
```
Do not copy placeholder identity values literally.

## 5. Git Log

### What it does
`git log` displays commit history, including commit hash, author, date, and message.

### Useful commands


````python
git log
git log --oneline
git log --oneline --graph --decorate --all
````

**Expected result:** Your commits appear newest first. `--oneline` makes the history compact; `--graph` shows branch/merge structure. The exact commits depend on your repository.

## 6. Git Diff

### What it does
`git diff` shows line-by-line changes that have not been staged. Use `git diff --staged` to inspect staged changes.

### Practical example
Edit `notes.txt`, then inspect both unstaged and staged changes:

````python
# PowerShell example
Add-Content -Path notes.txt -Value 'Practicing git diff'
git diff
git add notes.txt
git diff --staged
````

**Expected result:** The diff shows added/removed lines. Added lines begin with `+`; removed lines begin with `-`. After staging, ordinary `git diff` may be empty for that change, while `git diff --staged` shows it.

**Tip:** Run `git status` alongside diff to understand which area (working tree or staging area) contains the changes.

## 7. Git Show

### What it does
`git show` displays details about a Git object; by default it shows the latest commit and its patch.

### Useful commands


````python
git show
git show --stat
git show <commit-hash>
````

Replace `<commit-hash>` with a real hash from `git log --oneline`. Do not type the angle-bracket placeholder literally.

**Expected result:** Git shows the commit metadata and, normally, the file changes introduced by that commit.

## Troubleshooting quick reference

| Problem | What to check |
|---|---|
| `not a git repository` | Confirm the terminal is inside the intended repository. |
| Nothing appears in `git status` | The working tree may be clean, or you may be in a different folder. |
| `git commit` says nothing to commit | Check whether changes were staged with `git add`. |
| Wrong files staged | Inspect `git status`; unstage a file with `git restore --staged <file>`. |
| `git log` opens a pager | Press `q` to exit. |
