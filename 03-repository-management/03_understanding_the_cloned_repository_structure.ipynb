# 03. Understanding the Cloned Repository Structure

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Level | Beginner |

## 1. Learning Objectives
- Inspect visible and hidden repository files.
- Understand the role of the `.git` directory.
- Check the current branch and local repository configuration.

## 2. Inspect the Repository
Run from the cloned practice repository in PowerShell:
```powershell
Get-ChildItem -Force
Get-Content README.md
Get-ChildItem .git -Force
git branch
git config --local --list
git status
```

## 3. Understand the Output
- `Get-ChildItem -Force` includes hidden items such as `.git`.
- `README.md` is a normal project file.
- `.git` stores repository metadata and Git's object database, refs, configuration and other internal data.
- `git branch` lists local branches; `*` marks the current branch.
- `git config --local --list` displays repository-specific configuration.
- `git status` reports the current branch and working-tree state.

## 4. Important Warning
Do not manually delete or edit files inside `.git`. Doing so can damage the repository or make Git lose its metadata. Git manages this directory.

## 5. Common Mistakes
- Assuming `.git` is an ordinary project folder.
- Forgetting `-Force` when listing hidden items in PowerShell.
- Confusing local configuration with global Git configuration.

## 6. Exercise
Inspect a cloned repository and identify its README, `.git` directory, current branch and remote URL.

## 7. Interview Questions
**Q1. What is stored in `.git`?** Git's repository metadata, objects, refs and configuration.

**Q2. Why use `Get-ChildItem -Force`?** To include hidden items in PowerShell's directory listing.

**Q3. What does the asterisk in `git branch` mean?** It marks the currently checked-out branch.

## 8. Summary
A cloned repository contains project files plus a hidden `.git` directory that stores the information Git needs to manage history and repository state.
