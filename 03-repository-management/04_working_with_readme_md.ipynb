# 04. Working with README.md

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Level | Beginner |

## 1. Learning Objectives
- Understand the purpose of a README.
- Edit Markdown content in VS Code.
- Review, stage, commit and push a README update.

## 2. What is README.md?
`README.md` is a Markdown document that introduces a project. GitHub normally displays it on the repository's main page.

## 3. Example README
```markdown
# Git Practice Project

A repository for practicing Git and GitHub.

## Purpose

This repository is created to learn and practice Git and GitHub commands.

## Learning Goals

- Understand Git fundamentals.
- Practice repository management.
- Learn branching and merging.
- Understand commits and remote repositories.
```

## 4. Edit and Review
Open the file in VS Code:
```powershell
code README.md
```

After saving, inspect the change:
```powershell
git diff -- README.md
git status
```

## 5. Stage, Commit and Push
```powershell
git add README.md
git status
git commit -m "Update project README"
git status
git push origin main
```

Then refresh the repository page on GitHub to verify the updated README.

## 6. Common Mistakes
- Saving the file with the wrong extension, such as `README.md.txt`.
- Leaving accidental text or unwanted blank lines in the document.
- Forgetting to stage the file before committing.
- Forgetting to push the local commit to GitHub.

## 7. Exercise
Add a project overview and a short list of learning goals to a practice README. Review the diff before committing.

## 8. Interview Questions
**Q1. What does `.md` mean?** It is the common file extension for Markdown.

**Q2. Does editing README automatically update GitHub?** No. The change must be committed and pushed, or edited directly on GitHub.

**Q3. Which command previews unstaged README changes?** `git diff -- README.md`.

## 9. Summary
A clear README helps people understand a project. Use `git diff` to review edits, then stage, commit and push them.
