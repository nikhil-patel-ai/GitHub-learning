## 1. Learning Objectives
Define working directory; identify where edits happen; distinguish working-tree changes from staged and committed snapshots.

## 2. Key Concepts
Working directory (working tree) is the checked-out set of project files where you create, edit and delete files. Editing does not automatically stage or commit changes.

## 3. Key Terminology
Working tree: files currently checked out. Unstaged change: tracked file differs from index. Untracked file: new file not yet tracked. `git diff`: unstaged tracked-file differences.

## 7. Common Mistakes and Troubleshooting
Saving in VS Code does not stage a file. Do not edit the learning repository for practice; use the separate practice project. Check path first.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
