## 1. Learning Objectives
Define working directory/working tree; understand where edits happen; distinguish edits from staged and committed changes.

## 2. Key Concepts
The working directory (working tree) is the checked-out project files where you create, edit and remove files. Changes are not automatically staged or committed.

## 3. Commands and Practical Notes
`Set-Content working-demo.txt 'First version'`; `git status`; for a tracked file use `Add-Content README.md 'Working directory practice.'`; `git diff -- README.md`; `git status`.

## 4. Practical Exercise
Create or edit a file, inspect its working-tree state, and view an unstaged difference to a tracked file.

## 5. Common Mistakes and Safety
- Run commands from the intended practice repository and verify the path first.
- Inspect `git status` before changing repository state.
- Do not stage secrets, credentials or unrelated files.
- Treat command output as machine- and repository-state-dependent; do not assume exact output text.

## 6. Interview Questions
What is the working directory? Does saving a file automatically stage it?

## 7. Summary
Review the concept, run the relevant commands in the separate practice repository, and verify the resulting repository state before moving to the next topic.
