## 1. Learning Objectives
Define working directory; identify where edits happen; distinguish working-tree changes from staged and committed snapshots.

## 2. Key Concepts
The working directory (working tree) is the checked-out project files where you create, edit and delete files. Editing does not automatically stage or commit changes.

## 3. Key Terminology
Working tree: checked-out files. Unstaged change: tracked file differs from index. Untracked file: new file not yet tracked. `git diff`: unstaged tracked-file differences.

## 8. Best Practices
- Check the current path before running commands.
- Use `git status` before and after state-changing operations.
