# 05. Making Changes to Files and Tracking Them with Git

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Level | Beginner |

## 1. Learning Objectives
- Create a file in a Git repository.
- Recognize an untracked file.
- Stage changes and inspect the staging area.
- Commit and push tracked changes.

## 2. Git's Three Main Areas
- **Working directory:** Files currently being edited.
- **Staging area:** The changes selected for the next commit.
- **Repository:** The committed history.

## 3. Create a New File
Run in the practice repository:
```powershell
Set-Content -Path practice.txt -Value "Learning Git and GitHub."
Add-Content -Path practice.txt -Value "Practicing file tracking."
Get-Content practice.txt
git status
```

Git reports a new untracked file because it has not yet been staged.

## 4. Stage and Inspect
```powershell
git add practice.txt
git status
git diff --staged
```

`git add` stages the file. `git diff --staged` displays the staged content changes.

## 5. Commit the File
```powershell
git commit -m "Add practice text file"
git status
git log -1 --oneline
```

A commit records the staged snapshot in the local repository.

## 6. Modify the File
```powershell
Add-Content -Path practice.txt -Value "Learning how Git tracks modified files."
Get-Content practice.txt
git diff
git status
```

`git diff` displays unstaged modifications to tracked files.

## 7. Stage and Commit the Modification
```powershell
git add practice.txt
git diff --staged
git commit -m "Update practice text file"
git status
git log -2 --oneline
```

## 8. Push to GitHub
```powershell
git push origin main
git status
git log -3 --oneline
git remote -v
```

After pushing, open the GitHub repository and verify that `practice.txt` is present.

## 9. Common Mistakes
- Forgetting to stage a new file.
- Assuming `git add` creates a commit; it only stages changes.
- Assuming a local commit is already on GitHub; `git push` sends commits to the remote.

## 10. Interview Questions
**Q1. What is an untracked file?** A file Git sees in the working directory but is not yet tracking.

**Q2. What does `git add` do?** It stages selected changes for the next commit.

**Q3. What is the difference between `git diff` and `git diff --staged`?** The first shows unstaged changes; the second shows staged changes.

**Q4. Does `git commit` upload changes to GitHub?** No. It creates a local commit; use `git push` to upload it.

## 11. Summary
Git tracks a file after it is added to the staging area and committed. Review changes before committing, and push commits when you want them available on the remote repository.
