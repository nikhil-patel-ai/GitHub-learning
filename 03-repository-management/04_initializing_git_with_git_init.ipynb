## 1. Learning Objectives
Initialize a local repository; verify initialization; understand what the command does and does not do.

## 2. Key Concepts
`git init` creates Git metadata in the current directory. It does not create a commit or upload anything to GitHub.

## 3. Key Terminology
Initialize: create repository metadata. `.git`: internal Git data directory. Initial commit: first recorded snapshot, created separately.

## 4. Practical Work
Run commands from the intended practice folder. If it is already initialized, inspect status instead of repeating setup.

## 5. Verification and Expected Results
Confirm `git status` recognizes the repository and `Get-ChildItem -Force` shows `.git`. Exact hints vary by Git version/configuration.

**Actual output:** Add your own output after performing the exercise. No actual result is assumed here.

## 6. Common Mistakes and Troubleshooting
Check `Get-Location` before running `git init`. Do not initialize a parent or unrelated project.

## 7. Best Practices
- Confirm the current folder before running commands.
- Keep practice work separate from the learning-notes repository.
- Verify each operation instead of assuming it succeeded.

## 10. Summary
`git init` starts local version control in the current directory; commits and remote publishing are separate steps.
