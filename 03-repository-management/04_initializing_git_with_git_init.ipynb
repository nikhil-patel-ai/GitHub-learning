## 1. Learning Objectives
Run git init in the correct folder; verify initialization; understand what the command does not do.

## 2. Key Concepts
`git init` creates Git metadata in the current directory and initializes a repository. It does not create a commit or upload anything to GitHub.

## 3. Key Terminology
Initialize: create repository metadata. `.git`: internal Git data directory. Initial commit: first recorded snapshot, created separately.

## 7. Common Mistakes and Troubleshooting
Check `Get-Location` before running `git init`. Do not initialize a parent folder or an unrelated project by mistake.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
