## 1. Learning Objectives
Understand a hosted repository; create a practice repository; verify its name, visibility, README and default branch.

## 2. Key Concepts
A GitHub repository hosts project files and Git history online. The README introduces a project. The default branch is the branch GitHub opens by default.

## 3. Key Terminology
Repository: project files plus history. Public: generally viewable by anyone. Private: restricted to authorized users. README: project introduction. Default branch: default branch shown by GitHub.

## 4. Practical Work
Create a separate practice repository named `git-practice-project` on GitHub. Choose a description and visibility. Optionally initialize it with a README. Do not change the learning-notes repository.

## 5. Verification and Expected Results
Verify the repository URL, visibility, README, default branch and initial commit. Record actual values; do not assume example settings were selected.

## 6. Common Mistakes and Troubleshooting
Creating under the wrong account; unclear naming; publishing private information. Never commit passwords, tokens or API keys.

## 7. Best Practices
- Confirm the current folder before running commands.
- Keep practice work separate from the learning-notes repository.
- Verify each operation instead of assuming it succeeded.
