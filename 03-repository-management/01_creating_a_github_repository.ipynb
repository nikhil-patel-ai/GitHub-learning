## 1. Learning Objectives
Understand hosted repositories; create a repository; verify its name, visibility, README and default branch.

## 2. Key Concepts
A GitHub repository hosts project files and Git history online. It can be public or private. A README introduces the project. A default branch is the branch GitHub opens by default.

## 3. Key Terminology
Repository: project files plus history. Public: generally viewable by anyone. Private: restricted to authorized users. README: project introduction. Default branch: default branch shown by GitHub.

## 7. Common Mistakes and Troubleshooting
Common mistakes: creating under the wrong account or organization; unclear repository name; accidentally publishing private data. Never commit passwords, API keys, tokens or personal data.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
