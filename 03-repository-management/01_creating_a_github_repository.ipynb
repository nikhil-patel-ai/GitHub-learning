# 01. Creating a GitHub Repository

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Level | Beginner |

## 1. Learning Objectives
- Understand what a GitHub repository is.
- Create a repository with suitable name, description and visibility.
- Identify the README, default branch and initial commit.

## 2. Key Concepts
- **Repository:** A place to store project files and their revision history.
- **Public repository:** Visible to everyone.
- **Private repository:** Access is limited to people granted permission.
- **README.md:** A Markdown file commonly used to explain a project.
- **Default branch:** The branch GitHub opens by default (commonly `main`).

## 3. Create a Repository on GitHub
1. Sign in to GitHub and select **New repository**.
2. Enter a repository name and optional description.
3. Choose Public or Private.
4. Optionally select **Add a README file**.
5. Select **Create repository**.

## 4. Practical Work
For the learning exercise, create a separate practice repository rather than changing the learning-notes repository.

Example settings:
- Repository name: `git-practice-project`
- Description: `A repository for practicing Git and GitHub.`
- Visibility: Public
- Add README: Yes
- `.gitignore`: No .gitignore
- License: No license

## 5. Verify the Repository
Open the repository page and check the repository name, visibility, default branch, README and initial commit.

## 6. Common Mistakes
- Choosing a repository name that is difficult to understand.
- Accidentally creating the practice repository inside the wrong GitHub account or organization.
- Publishing passwords, API keys or private information in a public repository.

## 7. Exercise
Create a practice repository and write down its URL, visibility and default branch.

## 8. Interview Questions
**Q1. What is a GitHub repository?** A hosted project space containing files and Git history.

**Q2. What is the purpose of a README?** It introduces the project and explains how to use or understand it.

**Q3. Can a public repository be viewed by anyone?** Yes, unless access is restricted by other platform or organization policies.

## 9. Summary
A GitHub repository stores project files and their history online. Choose its name and visibility carefully, and use a README to explain the project.
