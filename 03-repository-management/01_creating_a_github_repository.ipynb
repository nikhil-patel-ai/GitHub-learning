## 1. Learning Objectives
Understand a hosted repository; create a practice repository; verify its name, visibility, README and default branch.

## 2. Key Concepts
A GitHub repository hosts project files and Git history online. The README introduces a project. The default branch is the branch GitHub opens by default.

## 3. Key Terminology
Repository: project files plus history. Public: generally viewable by anyone. Private: restricted to authorized users. README: project introduction. Default branch: default branch shown by GitHub.

## 4. Practical Work
Create a separate practice repository named `git-practice-project` on GitHub. Choose a description and visibility. Optionally initialize it with a README. Do not change the learning-notes repository.

## 6. Working with README.md

A `README.md` file introduces the project. GitHub normally displays it on the repository's main page. Markdown headings, paragraphs and lists make project information easy to scan.

Example content for the separate practice repository:

```markdown
# Git Practice Project

A repository for practicing Git and GitHub.

## Purpose
Learn and practice Git and GitHub commands.

## Learning Goals
- Understand Git fundamentals.
- Practice repository management.
- Learn branching and merging.
- Understand commits and remote repositories.
```

In VS Code, open the practice repository's README with `code README.md`. After saving, inspect it using `git diff -- README.md` and `git status`. To publish the update, stage, commit and push it (the push step assumes the remote and branch are configured):

```powershell
git add README.md
git status
git diff --staged
git commit -m "Update project README"
git push origin main
```

Refresh the GitHub repository page to verify the README. Editing a local README does not automatically update GitHub.

**Common mistakes:** saving as `README.md.txt`, forgetting to stage or commit, or forgetting to push. The `.md` extension indicates Markdown.
