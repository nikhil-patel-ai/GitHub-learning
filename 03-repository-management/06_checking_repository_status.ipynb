## 1. Learning Objectives
Use `git status`; recognize untracked, staged and clean states; verify state after operations.

## 2. Key Concepts
`git status` summarizes the current branch and state of the working tree and staging area. It does not prove local commits are pushed.

## 3. Key Terminology
Untracked: not tracked by Git. Staged: selected for next commit. Modified: tracked file changed. Clean: no pending changes reported.

## 7. Common Mistakes and Troubleshooting
A clean working tree does not mean commits are pushed. Inspect staged files and configure Git identity before first commit.

## 8. Best Practices
- Check the current path before running commands.
- Use `git status` before and after state-changing operations.
- Review changes before committing.
