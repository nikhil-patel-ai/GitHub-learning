## 1. Learning Objectives
Use `git status`; recognize untracked, staged and clean states; verify state after Git operations.

## 2. Key Concepts
`git status` summarizes current branch and the state of the working tree and staging area. It does not prove local commits are pushed.

## 3. Key Terminology
Untracked: not yet tracked by Git. Staged: selected for next commit. Modified: tracked file changed. Clean: no pending tracked/untracked changes reported.

## 7. Common Mistakes and Troubleshooting
Do not confuse clean working tree with synchronized remote. Inspect staged files before committing. Ensure Git identity is configured before first commit.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
