## 1. Learning Objectives
Use `git status`; recognize untracked, staged and clean states; verify repository state after operations.

## 2. Key Concepts
`git status` summarizes the current branch and the state of the working tree and staging area. It does not show whether local commits have been pushed.

## 3. Commands and Practical Notes
`git status`; `Set-Content status-demo.txt 'Status practice'`; `git status`; `git add status-demo.txt`; `git status`; `git commit -m 'Add status demonstration file'`; `git status`.

## 4. Practical Exercise
Observe status before staging, after staging and after committing. Record how the message changes.

## 5. Common Mistakes and Safety
- Run commands from the intended practice repository and verify the path first.
- Inspect `git status` before changing repository state.
- Do not stage secrets, credentials or unrelated files.
- Treat command output as machine- and repository-state-dependent; do not assume exact output text.

## 6. Interview Questions
What does `git status` show? Does a clean working tree prove changes are pushed?

## 7. Summary
Review the concept, run the relevant commands in the separate practice repository, and verify the resulting repository state before moving to the next topic.
