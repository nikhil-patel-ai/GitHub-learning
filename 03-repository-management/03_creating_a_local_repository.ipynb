## 1. Learning Objectives
Create a local project folder; understand local repositories; distinguish a folder from a Git repository.

## 2. Key Concepts
A local repository is stored on your computer. A normal folder is not automatically managed by Git; initialization is a separate step.

## 3. Key Terminology
Project directory: folder containing files. Local repository: project directory managed by Git. Working directory: checked-out files you edit.

## 4. Practical Work
Create a dedicated practice folder in PowerShell. If it already exists, inspect it instead of overwriting it.

## 5. Verification and Expected Results
Verify current path and README contents. Before initialization, `git status` is expected to say this is not a Git repository.
