## 1. Learning Objectives
Create a local project folder; understand local repository; distinguish folder from Git repository.

## 2. Key Concepts
A local repository is stored on your computer. A normal folder is not automatically managed by Git; Git initialization is a separate step.

## 3. Key Terminology
Project directory: folder containing project files. Local repository: project directory managed by Git. Working directory: checked-out files you edit.

## 7. Common Mistakes and Troubleshooting
Run commands in the intended directory. Avoid using an existing important project folder for experiments. Do not overwrite an existing README without checking it.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
