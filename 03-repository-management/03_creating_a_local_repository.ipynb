## 1. Learning Objectives

## 2. Key Concepts
A local repository is stored on your computer. A normal folder is not automatically managed by Git; initialization is a separate step.

## 3. Key Terminology
Project directory: folder containing files. Local repository: project directory managed by Git. Working directory: checked-out files you edit.

## 6. Cloning an Existing GitHub Repository

Creating a local repository with `git init` is one approach. If a repository already exists on GitHub, use `git clone` to create a local copy that includes its Git history and configures a remote, conventionally named `origin`.

```powershell
Set-Location "$HOME\NVSCode"
git clone https://github.com/nikhil-patel-ai/git-practice-project.git
Set-Location .\git-practice-project
git status
git remote -v
git log --oneline
```

Replace the URL with the actual repository URL you intend to clone. Make sure the destination folder does not already exist. Run Git commands from inside the cloned repository.

**Verification:** `git status` identifies the checked-out branch and working-tree state; `git remote -v` displays fetch/push URLs; `git log --oneline` displays available history.

**Key terms:** Clone = local copy of a remote repository; `origin` = conventional remote name; working tree = checked-out files you view and edit.

**Common mistakes:** incorrect URL, destination folder already exists, or running commands outside the cloned directory.
