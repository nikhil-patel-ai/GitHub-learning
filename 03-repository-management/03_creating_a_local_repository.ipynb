# creating a local repository

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Topic | Creating a Local Repository |
| Level | Beginner |

## 1. Learning Objectives
Understand local repositories; create a project folder; distinguish a normal folder from a Git repository.

## 2. Key Concepts
A local repository is a Git-managed project stored on your computer. A normal folder is not automatically a Git repository.

## 3. Commands and Practical Notes
PowerShell: `cd $HOME`; `New-Item -ItemType Directory -Name git-local-practice`; `Set-Location .\git-local-practice`; `Set-Content README.md '# Local Git Practice'`; `Get-Location`; `Get-ChildItem -Force`.

## 4. Practical Exercise
Create a dedicated practice folder and README. Inspect its path and contents. If the folder already exists, inspect it instead of overwriting it.

## 5. Common Mistakes and Safety
- Run commands from the intended practice repository and verify the path first.
- Inspect `git status` before changing repository state.
- Do not stage secrets, credentials or unrelated files.
- Treat command output as machine- and repository-state-dependent; do not assume exact output text.

## 6. Interview Questions
What is a local repository? Is every project folder automatically a Git repository?

## 7. Summary
Review the concept, run the relevant commands in the separate practice repository, and verify the resulting repository state before moving to the next topic.
