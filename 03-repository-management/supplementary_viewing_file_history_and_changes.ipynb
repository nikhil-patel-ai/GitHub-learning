## 1. Learning Objectives
- View the commit history of a repository.
- Understand commit details and the meaning of `HEAD`.
- Inspect changes made to a particular file.
- Compare commits and check whitespace errors.

## 2. Key Terminology
- **Commit:** A saved snapshot of tracked project changes.
- **HEAD:** A reference to the currently checked-out commit.
- **Parent commit:** The commit immediately before another commit.
- **Diff:** The difference between two versions of content.
- **Patch (`-p`):** A readable display of the changes introduced by commits.

## 3. View Repository History
Run these commands from inside your local practice repository:
```powershell
git log --oneline
git log --oneline --all --graph
git log -5
git status
```

| Command | Purpose |
|---|---|
| `git log --oneline` | Shows a compact commit list. |
| `git log --oneline --all --graph` | Shows commit history as a graph, including other refs. |
| `git log -5` | Shows details for the latest five commits (or fewer if fewer exist). |
| `git status` | Shows the current branch and working-tree state. |

## 4. Inspect a Commit
```powershell
git show HEAD
git show --stat HEAD
git show HEAD -- README.md
```

- `git show HEAD` displays the latest commit and its patch.
- `git show --stat HEAD` displays a summary of files and line changes.
- `git show HEAD -- README.md` displays the README changes included in that commit.

## 5. Inspect One File's History
```powershell
git log --oneline -- README.md
git log -p -- README.md
git show HEAD~1:README.md
git show HEAD:README.md
```

`HEAD~1` means the first parent of the current commit. The file lookup commands display the version of `README.md` stored in the selected commit. If the file did not exist at that commit, Git reports an error.

## 6. Compare Commits
```powershell
git diff HEAD~1 HEAD
git diff HEAD~1 HEAD -- README.md
git diff --check
git status
```

- `git diff HEAD~1 HEAD` compares the two commits.
- Adding `-- README.md` limits the comparison to that file.
- `git diff --check` checks the diff for whitespace errors.
- `git status` confirms whether the working tree has changes.

## 7. Important Notes
- These inspection commands do not change or create commits.
- `git log` normally shows history reachable from the current `HEAD`; adding `--all` includes other refs.
- `git diff` compares content; it does not replace the commit history shown by `git log`.
- If Git opens a pager and displays `(END)`, press `q` to exit.
- `HEAD~1` requires a commit with a parent. A repository with only one commit has no `HEAD~1`.

## 9. Interview Questions
**Q1. What does `git log` do?**  
It displays the commit history reachable from the selected revision.

**Q2. What is `HEAD`?**  
It identifies the currently checked-out commit (normally through the current branch).

**Q3. What is the difference between `git show` and `git diff`?**  
`git show` can display an object such as a commit and its changes; `git diff` compares two states of content.

**Q4. What does `HEAD~1` mean?**  
It refers to the first parent of the current commit.

## 10. Summary
Use `git log` to explore history, `git show` to inspect commits, and `git diff` to compare versions. These commands help understand what changed, when it changed, and which commit introduced the change.
