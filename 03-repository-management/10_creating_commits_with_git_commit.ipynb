## 1. Learning Objectives
Understand commit snapshots; commit staged changes; verify commit locally.

## 2. Key Concepts
A commit records the staged snapshot in local history. It includes an identifier, author metadata, timestamp, message and parent relationship. It does not push to GitHub.

## 3. Key Terminology
Commit: saved snapshot of staged content. Commit ID: hash identifying a commit. `HEAD`: currently checked-out commit/reference. `git show`: inspect commit details.

## 7. Common Mistakes and Troubleshooting
Do not commit secrets or unrelated files. Configure user.name and user.email. Write a meaningful message. A commit is not a push.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
