# creating commits with git commit

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Topic | Creating Commits with git commit |
| Level | Beginner |

## 1. Learning Objectives
Understand commits; commit staged changes; verify the resulting local commit.

## 2. Key Concepts
A commit records a snapshot of staged changes in local repository history. It does not automatically push to GitHub.

## 3. Commands and Practical Notes
`git status`; `git diff --staged`; `git commit -m 'Add repository practice example'`; `git status`; `git log -1 --oneline`; `git show --stat HEAD`.

## 4. Practical Exercise
Make an intentional change, stage it, review the staged diff, commit it and verify the latest commit.

## 5. Common Mistakes and Safety
- Run commands from the intended practice repository and verify the path first.
- Inspect `git status` before changing repository state.
- Do not stage secrets, credentials or unrelated files.
- Treat command output as machine- and repository-state-dependent; do not assume exact output text.

## 6. Interview Questions
What does `git commit` do? Does committing upload changes to GitHub?

## 7. Summary
Review the concept, run the relevant commands in the separate practice repository, and verify the resulting repository state before moving to the next topic.
