## 1. Learning Objectives
Understand commit snapshots; commit staged changes; verify local history.

## 2. Key Concepts
A commit records the staged snapshot in local history. It includes an identifier, author metadata, timestamp, message and parent relationship. It does not push to GitHub.

## 3. Key Terminology
Commit: saved snapshot of staged content. Commit ID: hash identifying a commit. `HEAD`: current checked-out commit/reference. `git show`: inspect a commit.

## 7. Common Mistakes and Troubleshooting
Do not commit secrets or unrelated files. Configure `user.name` and `user.email`. A commit is not a push.

## 8. Best Practices
- Confirm the current repository path.
- Inspect `git status` before and after state-changing commands.
- Review staged content before committing.
- Keep secrets and unrelated files out of commits.

## 6. Modify, Commit and Push a Tracked File

A local commit does not automatically appear on GitHub. Push commits to the configured remote when you want to publish them.

```powershell
git commit -m "Add practice text file"
git status
git log -1 --oneline
Add-Content -Path practice.txt -Value "Learning how Git tracks modified files."
git diff -- practice.txt
git add practice.txt
git diff --staged
git commit -m "Update practice text file"
git status
git log -2 --oneline
git push origin main
git status
git remote -v
```

Run the first commit only when `practice.txt` is staged and ready. The push command assumes your remote is named `origin` and the branch is `main`; verify with `git remote -v` and `git branch --show-current` before pushing. Refresh GitHub to confirm the file is present.

## 7. Undoing Changes Safely

Before undoing anything, inspect the repository and protect work you may need:

```powershell
git status
git diff
git diff --staged
```

- **Unstage but keep edits:** `git restore --staged <file>` removes the file from the index while keeping its working-tree edits.
- **Discard unstaged edits:** `git restore <file>` replaces the working copy with the index version. This discards unstaged edits to that file.
- **Restore from latest commit:** `git restore --source=HEAD --staged --worktree <file>` resets both index and working copy for that path to `HEAD`; staged and unstaged edits to that file can be lost.
- **Reverse a commit:** `git revert <commit-hash>` creates a new commit that reverses an earlier commit and preserves history. Use the actual commit ID from `git log --oneline`; Git may open an editor to confirm the message.

### Understanding git reset

- `git reset --soft <commit>`: moves the current branch while keeping changes staged.
- `git reset --mixed <commit>`: moves the branch and leaves changes in the working tree, unstaged (default mode).
- `git reset --hard <commit>`: resets branch, index and working tree; can permanently discard uncommitted work. Do not experiment with this in an important repository.

### Understanding git clean

`git clean` removes untracked files/directories. Preview first with `git clean -n`. Do not run `git clean -f` or `git clean -fd` unless you have reviewed the preview and are certain the files are disposable.

| Command | Purpose |
|---|---|
| `git restore --staged <file>` | Unstage; keep working copy edits |
| `git restore <file>` | Discard unstaged edits to a tracked file |
| `git revert <commit>` | Add a new commit reversing an earlier commit |
| `git reset --soft <commit>` | Move branch; keep changes staged |
| `git reset --mixed <commit>` | Move branch; keep changes unstaged |
| `git reset --hard <commit>` | Reset index and working tree; destructive |
| `git clean -n` | Preview untracked-file removal |

**Safety rule:** Never run a destructive undo command until you understand which changes it affects. Back up valuable work first. Prefer `git revert` when reversing a commit that has already been shared.

**Practice:** Use a disposable test file. Practice unstaging with `git restore --staged`; practice `git revert` only against a commit you intentionally created and understand. Do not practice `git reset --hard` in your learning repository.

**Interview questions:** How do you unstage but keep edits? How do `git revert` and `git reset` differ? Which reset mode is destructive? How do you preview `git clean`?
