## 1. Learning Objectives
Understand commit snapshots; commit staged changes; verify local history.

## 2. Key Concepts
A commit records the staged snapshot in local history. It includes an identifier, author metadata, timestamp, message and parent relationship. It does not push to GitHub.

## 3. Key Terminology
Commit: saved snapshot of staged content. Commit ID: hash identifying a commit. `HEAD`: current checked-out commit/reference. `git show`: inspect a commit.

## 4. Commands and Syntax
Run commands from the intended practice repository. Review comments and status before executing state-changing commands.

## 5. Practical Work
Check status and staged diff; commit an intentional staged change; verify with `git log` and `git show`.

## 6. Verification and Expected Results
A successful commit appears in local history. If Git says nothing to commit, check whether changes were made and staged.

**Actual output:** Record your own terminal output after performing the exercise. No actual output is assumed in this notebook.

## 7. Common Mistakes and Troubleshooting
Do not commit secrets or unrelated files. Configure `user.name` and `user.email`. A commit is not a push.

## 8. Best Practices
- Confirm the current repository path.
- Inspect `git status` before and after state-changing commands.
- Review staged content before committing.
- Keep secrets and unrelated files out of commits.

## 9. Exercise
Make a small change, stage it, inspect it, commit it and verify the latest commit.

## 10. Interview Questions
What does `git commit` do? Does it upload changes? Why review staged changes first?

## 11. Summary
A commit records staged changes in local history. Review, commit and verify; pushing is separate.
