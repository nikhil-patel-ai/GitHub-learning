## 1. Learning Objectives
Stage new files and tracked modifications; verify what is staged.

## 2. Key Concepts
`git add` copies the current file content into the index for the next commit. It does not create a commit or upload to GitHub.

## 3. Key Terminology
`git add <file>`: stage a path. `git add .`: stage eligible changes under current directory. `git diff --staged`: review staged changes.

## 7. Common Mistakes and Troubleshooting
`git add .` can stage more than intended. Check status first and inspect staged diff. Never stage credentials or private data.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
