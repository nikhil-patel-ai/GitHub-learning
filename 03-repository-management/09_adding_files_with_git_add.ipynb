## 1. Learning Objectives
Stage new files and tracked modifications; verify staged content.

## 2. Key Concepts
`git add` places selected file content into the index for the next commit. It does not create a commit or upload to GitHub.

## 3. Key Terminology
`git add <file>` stages a path. `git add .` stages eligible changes under the current directory. `git diff --staged` reviews staged changes.

## 4. Commands and Syntax
Run commands from the intended practice repository. Review comments and status before executing state-changing commands.

## 5. Practical Work

## 6. Verification and Expected Results

## 7. Common Mistakes and Troubleshooting
`git add .` can stage more than intended. Inspect status first. Never stage credentials or private data.

## 8. Best Practices
- Confirm the current repository path.
- Inspect `git status` before and after state-changing commands.
- Review staged content before committing.
- Keep secrets and unrelated files out of commits.
