## 1. Learning Objectives
Understand index/staging area; distinguish staged and unstaged changes; review next commit content.

## 2. Key Concepts
The staging area (index) holds the selected snapshot of changes for the next commit. Staged and unstaged changes can coexist.

## 3. Key Terminology
Index: another name for staging area. Staged change: selected for next commit. Unstaged change: working-tree change not yet staged. `git diff --staged`: inspect staged changes.

Use PowerShell or Git Bash from the intended practice repository. Commands are examples; run only the commands relevant to your current exercise.

## 6. Verification and Expected Results
The staged diff shows the content selected at `git add` time; the unstaged diff shows subsequent edits. Exact output depends on your file contents.

## 7. Common Mistakes and Troubleshooting
Do not assume later edits are automatically staged. Review `git diff --staged` before committing. Avoid staging secrets.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
