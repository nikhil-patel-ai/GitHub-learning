## 1. Learning Objectives
Understand the index/staging area; distinguish staged and unstaged changes; review the next commit content.

## 2. Key Concepts
The staging area (index) holds the selected snapshot of changes for the next commit. Staged and unstaged changes can coexist.

## 3. Key Terminology
Index: another name for staging area. Staged change: selected for next commit. Unstaged change: working-tree change not yet staged. `git diff --staged`: inspect staged changes.

## 7. Common Mistakes and Troubleshooting
Do not assume later edits are automatically staged. Review `git diff --staged` before committing and avoid staging secrets.

## 8. Best Practices
- Check the current path before running commands.
- Use `git status` before and after state-changing operations.
- Review changes before committing.

## 6. Stage and Review a File

The staging area lets you choose exactly which changes will be included in the next commit.

```powershell
git add practice.txt
git status
git diff --staged
```

`git add` stages the current file content. `git diff --staged` displays the staged changes. Staged and unstaged changes can coexist if you edit the file again after staging it.
