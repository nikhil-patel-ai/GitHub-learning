## 1. Learning Objectives
View history; interpret commit IDs, authors and messages; inspect commits and file history.

## 2. Key Concepts
`git log` displays reachable commit history. `git show` inspects a commit. `HEAD` refers to the current checked-out commit, usually through the current branch.

## 3. Key Terminology
Commit history: recorded snapshots. `--oneline`: compact display. `--graph`: ASCII graph. `HEAD`: current checkout reference.

## 4. Commands and Syntax
Run commands from the intended practice repository. Review comments and status before executing state-changing commands.

## 5. Practical Work

## 6. Verification and Expected Results

## 7. Common Mistakes and Troubleshooting
Uncommitted changes do not appear as commits in `git log`. History depends on current branch and reachable refs.

## 8. Best Practices
- Confirm the current repository path.
- Inspect `git status` before and after state-changing commands.
- Review staged content before committing.
- Keep secrets and unrelated files out of commits.

## 11. Summary
Use `git log` to browse history and `git show` to inspect a specific commit.

## 6. Detailed History, File History and Commit Comparison

Use these commands from the local practice repository:

```powershell
git log --oneline
git log --oneline --all --graph
git log -5
git status
git show HEAD
git show --stat HEAD
git show HEAD -- README.md
git log --oneline -- README.md
git log -p -- README.md
git show HEAD~1:README.md
git show HEAD:README.md
git diff HEAD~1 HEAD
git diff HEAD~1 HEAD -- README.md
git diff --check
git status
```

- `git log --oneline --all --graph` shows a compact graph including other refs.
- `git show --stat HEAD` summarizes files and line changes in the latest commit.
- `git log -p -- README.md` displays patches affecting that file.
- `HEAD~1` means the first parent of the current commit. It requires a commit with a parent; a one-commit repository has no `HEAD~1`.
- `git show HEAD:README.md` displays the README version stored in the current commit. Git reports an error if the file did not exist at that commit.
- `git diff HEAD~1 HEAD` compares two commits; adding `-- README.md` limits the comparison to that file.
- `git diff --check` checks the diff for whitespace errors.

These are inspection commands; they do not create commits. `git log` normally shows history reachable from the current `HEAD`; `--all` includes other refs. If Git opens a pager and displays `(END)`, press `q` to exit.
