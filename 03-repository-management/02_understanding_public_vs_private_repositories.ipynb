# understanding public vs private repositories

| Field | Details |

| Level | Beginner |

## 1. Learning Objectives
Distinguish public and private repository visibility; understand access implications; choose suitable visibility.

## 2. Key Concepts
A public repository can generally be viewed by anyone. A private repository is limited to its owner and authorized users. Private visibility is not a substitute for secret management.

## 3. Commands and Practical Notes
Inspect the visibility label on a practice repository. Review Settings without changing visibility. Never commit passwords, tokens, API keys or private data.

## 4. Practical Exercise
Inspect a practice repository and explain why its visibility setting is appropriate.

## 5. Common Mistakes and Safety
- Run commands from the intended practice repository and verify the path first.
- Inspect `git status` before changing repository state.
- Do not stage secrets, credentials or unrelated files.
- Treat command output as machine- and repository-state-dependent; do not assume exact output text.

## 6. Interview Questions
What is the main difference between public and private repositories? Can a private repository safely store API keys?

## 7. Summary
Review the concept, run the relevant commands in the separate practice repository, and verify the resulting repository state before moving to the next topic.
