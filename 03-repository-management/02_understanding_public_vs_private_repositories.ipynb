## 1. Learning Objectives
Distinguish visibility; understand access; choose suitable visibility.

## 2. Key Concepts
Public repositories can generally be viewed by anyone. Private repositories are limited to the owner and authorized users. Private visibility does not make committing secrets safe.

## 3. Key Terminology
Visibility: who can access content. Collaborator: person granted access. Public: broadly viewable. Private: access-controlled.

## 4. Practical Work
Open a practice repository, locate its visibility label, and review Settings to understand where visibility is managed. Do not change visibility unless intended.

## 5. Verification and Expected Results
Record current visibility and explain why it suits the practice project. Do not assume a private repository is a secret manager.

## 6. Common Mistakes and Troubleshooting
Do not publish credentials or confidential information. Check organization policies before changing visibility.

## 7. Best Practices
- Confirm the current folder before running commands.
- Keep practice work separate from the learning-notes repository.
- Verify each operation instead of assuming it succeeded.

## 8. Exercise
Compare public and private repository access and list who can view each.

## 9. Interview Questions
What is the difference between public and private repositories? Can a private repository safely store API keys?

## 10. Summary
Visibility determines who can view repository content. Choose it carefully and keep secrets out of Git history.
