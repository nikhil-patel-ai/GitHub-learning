## 1. Learning Objectives
Distinguish repository visibility; understand access; choose suitable visibility.

## 2. Key Concepts
Public repositories can generally be viewed by anyone. Private repositories are limited to the owner and authorized users. Private visibility does not make committing secrets safe.

## 3. Key Terminology
Visibility: who can access repository content. Collaborator: user granted access. Public: discoverable and viewable broadly. Private: access controlled.

## 7. Common Mistakes and Troubleshooting
Do not publish credentials or confidential information. Check account/organization policies before changing visibility. Visibility changes may affect who can access project content.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
