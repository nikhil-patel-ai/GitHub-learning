# 02. Cloning a GitHub Repository

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Level | Beginner |

## 1. Learning Objectives
- Understand what cloning does.
- Clone a remote repository to a local computer.
- Verify the cloned repository and its remote.

## 2. Key Concepts
- **Clone:** Creates a local copy of a remote repository, including its Git history.
- **origin:** The conventional name assigned to the remote from which a repository was cloned.
- **Working tree:** The checked-out files that you view and edit.

## 3. Clone a Repository
Run these commands in PowerShell:
```powershell
cd "C:\Users\Nikhil\NVSCode"
git clone https://github.com/nikhil-patel-ai/git-practice-project.git
cd .\git-practice-project
git status
```

## 4. Verify the Remote and History
```powershell
git remote -v
git log --oneline
```

Expected checks:
- `git status` identifies the current branch and reports whether the working tree is clean.
- `git remote -v` shows fetch and push URLs for `origin`.
- `git log --oneline` displays the available commit history.

## 5. Common Mistakes
- Cloning into a folder where a directory with the same name already exists.
- Copying an incorrect repository URL.
- Running repository commands from a directory that is not inside the cloned repository.

## 6. Exercise
Clone another public practice repository into a separate folder and inspect its remote URL.

## 7. Interview Questions
**Q1. What does `git clone` do?** It creates a local repository copy from a remote source.

**Q2. What is `origin`?** It is the conventional remote name created by `git clone`.

**Q3. Does cloning include commit history?** Yes, the clone normally includes the repository history needed for its branches and refs.

## 8. Summary
`git clone URL` downloads a repository into a new local directory and configures a remote, conventionally named `origin`.
