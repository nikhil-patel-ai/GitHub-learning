## 2. Key Concepts
`.git` stores repository metadata, objects, references and configuration. Its exact layout varies with Git version and repository state.

## 3. Key Terminology
`config`: repository configuration. `HEAD`: current checkout reference/commit pointer. `objects`: Git object database. `refs`: branch/tag references. `hooks`: hook scripts.

## 6. Inspecting a Cloned Repository

A cloned repository contains normal project files and the hidden `.git` directory. Use these read-only commands from the cloned repository root:

```powershell
Get-ChildItem -Force
Get-Content README.md
Get-ChildItem .git -Force
git branch
git config --local --list
git status
git remote -v
```

- `Get-ChildItem -Force` includes hidden entries such as `.git`.
- `README.md` is an ordinary project file.
- `git branch` lists local branches; `*` marks the current branch.
- `git config --local --list` displays repository-specific configuration.
- `git remote -v` shows configured remote URLs.

**Important:** Do not manually delete or edit files inside `.git`; doing so can damage repository metadata. Use Git commands to manage the repository.

**Common mistakes:** treating `.git` as an ordinary project folder, forgetting `-Force` in PowerShell, and confusing local configuration with global configuration.
