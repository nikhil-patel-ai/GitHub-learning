## 1. Learning Objectives

## 2. Key Concepts
`.git` stores repository metadata, objects, references and configuration. Its exact layout varies with Git version and repository state.

## 3. Key Terminology
`config`: repository configuration. `HEAD`: current checkout reference/commit pointer. `objects`: Git object database. `refs`: branch/tag references. `hooks`: hook scripts.

## 7. Common Mistakes and Troubleshooting
Never manually delete, rename or edit `.git` internals. Use Git commands to manage repository state.

## 8. Best Practices
- Check the current path before running commands.
- Use `git status` before and after state-changing operations.
- Review changes before committing.
