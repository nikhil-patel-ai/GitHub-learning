# understanding the git directory

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Topic | Understanding the .git Directory |
| Level | Beginner |

## 1. Learning Objectives
Locate `.git`; understand its role; inspect its contents safely.

## 2. Key Concepts
The hidden `.git` directory stores repository metadata, Git objects, references and configuration. Its internal layout can vary by Git version and repository state.

## 3. Commands and Practical Notes
PowerShell: `Get-ChildItem -Force`; `Get-ChildItem .git -Force`; `Get-Content .git\config`. Do not edit or delete internal files.

## 4. Practical Exercise
Locate `.git` and inspect its contents without modifying anything.

## 5. Common Mistakes and Safety
- Run commands from the intended practice repository and verify the path first.
- Inspect `git status` before changing repository state.
- Do not stage secrets, credentials or unrelated files.
- Treat command output as machine- and repository-state-dependent; do not assume exact output text.

## 6. Interview Questions
What is stored in `.git`? Why should you not manually delete or edit it?

## 7. Summary
Review the concept, run the relevant commands in the separate practice repository, and verify the resulting repository state before moving to the next topic.
