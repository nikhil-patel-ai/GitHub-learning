## 1. Learning Objectives
Locate `.git`; identify common metadata; inspect safely.

## 2. Key Concepts
`.git` stores repository metadata, object data, references and configuration. Its exact internal layout may vary with Git version and repository state.

## 3. Key Terminology
`config`: repository configuration. `HEAD`: current checked-out reference/commit pointer. `objects`: Git object database. `refs`: references such as branches/tags. `hooks`: hook scripts.

## 7. Common Mistakes and Troubleshooting
Do not manually delete, rename or edit `.git` contents. Use Git commands to manage repository state. `-Force` shows hidden entries in PowerShell.

## 8. Best Practices
- Confirm the current repository path before running commands.
- Review `git status` before changing repository state.
- Keep practice work in the separate practice repository.
- Verify each operation instead of assuming it succeeded.
