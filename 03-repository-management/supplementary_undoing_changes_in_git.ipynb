# 07. Undoing Changes in Git

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Level | Beginner |

## 1. Learning Objectives
- Discard unstaged changes safely.
- Unstage files without deleting their contents.
- Understand the difference between `restore`, `reset` and `revert`.
- Recognize commands that can permanently remove uncommitted work.

## 2. Important Safety Rule
Before undoing anything, inspect the repository:
```powershell
git status
git diff
git diff --staged
```
Do not run destructive commands until you understand which changes will be affected. If the work matters, make a backup or commit it first.

## 3. Unstage a File with git restore --staged
If a file is staged but you want to remove it from the staging area while keeping its edits:
```powershell
git restore --staged README.md
git status
```
This changes the staging area; it does not discard the file's working-directory edits.

## 4. Discard Unstaged Changes with git restore
To replace the working copy of a tracked file with the version from the index:
```powershell
git diff -- README.md
git restore README.md
git status
```
**Warning:** This discards unstaged edits to that file. Make sure you do not need them.

## 5. Restore a File from the Latest Commit
```powershell
git restore --source=HEAD --staged --worktree README.md
git status
```
This restores both the index and working-tree copy of the file from `HEAD`. It can discard staged and unstaged changes to that file.

## 6. Undo a Commit with git revert
`git revert` creates a new commit that reverses the changes introduced by an earlier commit. It preserves the existing history.
```powershell
git log --oneline
git revert <commit-hash>
git log --oneline
```
Replace `<commit-hash>` with the actual commit ID. Git may open an editor to confirm the revert message. Do not type the angle brackets.

## 7. Understand git reset (Concept Only)
- `git reset --soft <commit>` moves the current branch while keeping changes staged.
- `git reset --mixed <commit>` (the default mode) moves the branch and keeps changes in the working tree, unstaged.
- `git reset --hard <commit>` moves the branch and resets the index and working tree. It can permanently discard work.

Do not practice `git reset --hard` on your learning repository. Use a disposable test repository only after understanding the consequences.

## 8. Understand git clean (Concept Only)
`git clean` removes untracked files and directories. It is destructive and does not use Git's normal tracked-file history.
Preview first:
```powershell
git clean -n
```
Do not run `git clean -f` or `git clean -fd` in a real project unless you have checked the preview and are certain the untracked files are disposable.

## 9. Command Comparison
| Command | Main purpose |
|---|---|
| `git restore --staged <file>` | Unstage a file, keeping its working copy edits. |
| `git restore <file>` | Discard unstaged edits to a tracked file. |
| `git revert <commit>` | Create a new commit that reverses an earlier commit. |
| `git reset --soft <commit>` | Move branch; keep changes staged. |
| `git reset --mixed <commit>` | Move branch; keep changes unstaged. |
| `git reset --hard <commit>` | Move branch and discard index/working-tree changes. |
| `git clean -n` | Preview removal of untracked files. |

## 10. Practical Exercise
Use a disposable test file in your practice repository. First inspect status and diffs, then practice unstaging with `git restore --staged`. Do not discard any real work. Practice `git revert` only against a commit you intentionally created and understand.

## 11. Interview Questions
**Q1. How do you unstage a file but keep its edits?** Use `git restore --staged <file>`.

**Q2. What is the difference between `git revert` and `git reset`?** Revert adds a new commit reversing changes; reset moves the current branch reference and may alter the index or working tree.

**Q3. Which reset mode is destructive?** `--hard` resets both the index and working tree and can discard uncommitted work.

**Q4. How can you preview files that `git clean` would remove?** Use `git clean -n`.

## 12. Summary
Choose the undo command based on where the change exists: working directory, staging area or commit history. Inspect first, protect valuable work, and prefer `git revert` when reversing published history.
