# understanding commit messages

| Field | Details |
|---|---|
| Module | 03 – Repository Management |
| Topic | Understanding Commit Messages |
| Level | Beginner |

## 1. Learning Objectives
Understand the purpose of commit messages; write concise and specific messages; recognize vague messages.

## 2. Key Concepts
A commit message explains the intent of a change. Good messages are concise, specific and accurate. Teams should use a consistent convention.

## 3. Commands and Practical Notes
After staging a real change: `git status`; `git diff --staged`; `git commit -m 'Add repository setup example'`; `git log -1 --oneline`. Choose a message that accurately describes your own change.

## 4. Practical Exercise
Create two small commits with distinct, meaningful messages and review them using `git log --oneline`.

## 5. Common Mistakes and Safety
- Run commands from the intended practice repository and verify the path first.
- Inspect `git status` before changing repository state.
- Do not stage secrets, credentials or unrelated files.
- Treat command output as machine- and repository-state-dependent; do not assume exact output text.

## 6. Interview Questions
Why are commit messages important? What makes a commit message useful?

## 7. Summary
Review the concept, run the relevant commands in the separate practice repository, and verify the resulting repository state before moving to the next topic.
