## 1. Learning Objectives
Explain commit-message purpose; write clear messages; distinguish subject and body.

## 2. Key Concepts
A commit message explains the intent of a change. A useful subject is concise, specific and accurate. A longer body can add context.

## 3. Key Terminology
Subject: short summary line. Body: optional explanation. Convention: consistent style followed by a project/team. Conventional Commits is covered later in Professional Git Workflow.

## 4. Commands and Syntax
Run commands from the intended practice repository. Review comments and status before executing state-changing commands.

## 5. Practical Work
After staging a real change, inspect it, commit with a message that accurately describes your change, and review the latest log entry.

## 6. Verification and Expected Results
Create two small commits with distinct messages. Review using `git log --oneline`. Example text is illustrative, not a claim about your actual change.

**Actual output:** Record your own terminal output after performing the exercise. No actual output is assumed in this notebook.

## 7. Common Mistakes and Troubleshooting
Avoid vague messages like `update`, `changes`, `final`. Do not claim a change that was not made. Follow team conventions.

## 8. Best Practices
- Confirm the current repository path.
- Inspect `git status` before and after state-changing commands.
- Review staged content before committing.
- Keep secrets and unrelated files out of commits.
