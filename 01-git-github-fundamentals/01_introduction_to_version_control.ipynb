## 03. What Is Version Control?

**Version control** is a system that records changes made to files over time. It helps developers review earlier versions, understand what changed, and work safely without losing previous work.

For example, while developing the **EduSphere LMS**, you may create a login page, update its design, and later fix a login bug. Version control helps track these changes so you can review the project history and return to an earlier version when needed.

A version control system can track source code and other project files.

## 04. Benefits of Version Control

1. **Change history:** Records changes made to project files.
2. **Recovery:** Helps restore an earlier version when necessary.
3. **Collaboration:** Allows multiple developers to work on a project.
4. **Accountability:** Helps identify which changes were made and when.
5. **Branching:** Allows developers to work on separate features without directly changing the main line of work.
6. **Safer development:** Makes it easier to experiment and review changes.

## 05. What Is Git?

**Git** is a distributed version control system. It tracks changes in files and stores project history in a repository.

Important points:

- Git is installed and used on a computer through a terminal or a graphical application.
- A Git repository stores project files and their version history.
- Git supports commits and branches.
- Git can be used without GitHub or another hosting service.

Git is the version control tool; GitHub is one service that can host Git repositories online.

## 06. What Is GitHub?

**GitHub** is an online platform for hosting and collaborating on Git repositories.

It provides features such as:

- Remote repository hosting.
- Pull requests for proposing and reviewing changes.
- Issues for tracking bugs and tasks.
- Collaboration tools for teams.
- Project visibility and sharing.

GitHub uses Git, but Git and GitHub are not the same thing.

## 07. Git vs GitHub

| Git | GitHub |
|---|---|
| Version control system | Online repository hosting and collaboration platform |
| Tracks changes and project history | Hosts Git repositories remotely |
| Can work locally without internet | Internet is generally needed to access the hosted service |
| Used through commands or Git clients | Used through a website, integrations, and Git clients |

**Remember:** Git manages version history. GitHub helps store and collaborate on Git repositories online.

## 08. Understanding Repositories: Local vs Remote

A **repository (repo)** is a project location managed by Git. It contains project files and version history.

- **Local repository:** The Git repository stored on your own computer.
- **Remote repository:** A repository hosted elsewhere, commonly on a platform such as GitHub.

A developer can make and commit changes locally, then push those commits to a remote repository. A remote repository can also be used to share work with collaborators.

## 09. Important Terminology

| Term | Meaning |
|---|---|
| Repository | A project managed by Git, including its history |
| Commit | A recorded snapshot of changes |
| Commit history | The sequence of commits in a repository |
| Working directory | The checked-out project files you are currently working on |
| Staging area | The area where selected changes are prepared for the next commit |
| Branch | An independent line of development |
| HEAD | A reference to the currently checked-out commit or branch position |
| Clone | Create a local copy of a remote repository |
| Push | Send local commits to a remote repository |
| Pull | Fetch remote changes and integrate them into the current branch |
| Fetch | Download remote updates without integrating them into the current branch |

## 10. Basic Git Workflow

A simplified workflow:

```text
Working Directory
       |
       | git add
       v
  Staging Area
       |
       | git commit
       v
  Local Repository
       |
       | git push
       v
  Remote Repository (for example, GitHub)
```

1. Edit files in the working directory.
2. Select changes for the next commit using `git add`.
3. Record the staged changes using `git commit`.
4. Share local commits with a remote repository using `git push`.

This is a conceptual workflow. The five commands practiced in this topic are inspection and history commands; they do not all perform these workflow transitions.

## 11. Basic Git Commands – Five Practical Exercises

The following five commands were part of the completed practice record.

### 1. Check the installed Git version

```bash
git --version
```

**Purpose:** Displays the installed Git version.  
**Expected/sample output:**

```text
git version 2.x.x
```

The exact version depends on the Git installation.

### 2. Check repository status

```bash
git status
```

**Purpose:** Shows the current branch and the state of the working directory and staging area.  
**Expected/sample output (when the working tree is clean):**

```text
On branch main
nothing to commit, working tree clean
```

The exact message depends on the repository state and branch name.

### 3. View compact commit history

```bash
git log --oneline
```

**Purpose:** Displays commit history in a compact, one-line-per-commit format.  
**Expected/sample output:**

```text
a1b2c3d Update login page
e4f5g6h Initial commit
```

The hashes and messages above are illustrative, not the user's captured terminal output.

### 4. View unstaged changes

```bash
git diff
```

**Purpose:** Displays differences in tracked files that have not been staged.

**Practice result recorded:** Command completed successfully. No output was expected because there were no unstaged changes at the time of practice.

### 5. View the latest commit summary

```bash
git show --stat
```

**Purpose:** Shows details of the selected commit, including a summary of files changed and line changes.

**Expected/sample output (illustrative):**

```text
commit <commit-hash>
Author: <author>
Date:   <date>

    Update project files

 path/to/file.py | 10 +++++-----
 1 file changed, 5 insertions(+), 5 deletions(-)
```

The exact output depends on the latest commit in the repository.

## 13. Troubleshooting

| Situation | What to check |
|---|---|
| `git` is not recognized | Check whether Git is installed and whether its executable is available in the system PATH. Reopen the terminal after installation or PATH changes. |
| `git status` says this is not a Git repository | Confirm that the terminal is inside the intended repository folder. |
| `git log` shows no commits | The repository may not have any commits yet. |
| `git diff` shows no output | There may be no unstaged changes in tracked files. This can be a normal result. |
| `git show --stat` cannot show a commit | Check that the repository contains at least one commit and that the command is being run inside the repository. |

Do not treat an empty `git diff` result as an error by itself.
