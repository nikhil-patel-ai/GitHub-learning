## 1. Topic Introduction

Git and GitHub are closely related but different. **Git** is a distributed version control system that tracks changes. **GitHub** is an online platform that hosts Git repositories and provides collaboration tools.

## 3. What Is Git?

Git is a distributed version control system. It records file changes as commits, supports branches, and lets developers review project history. Local Git operations can work offline.

Examples: `git status`, `git add`, `git commit`, and `git log`.

---

## 4. What Is GitHub?

GitHub is an online platform for hosting Git repositories. It adds collaboration features such as Pull Requests, Issues, Discussions, Projects, and GitHub Actions. Git itself does not require GitHub.

---

## 5. Git vs GitHub

| Git | GitHub |
|---|---|
| Distributed version control system | Online hosting and collaboration platform |
| Tracks file changes | Hosts remote Git repositories |
| Works locally and offline for local operations | Online features require internet |
| Records commits and branches | Adds Pull Requests, Issues, reviews, permissions, and automation |

**Remember:** Git tracks versions; GitHub is one service that hosts Git repositories.

---

## 6. Local Repository vs Remote Repository

**Local repository:** Exists on your computer, contains files and Git history, and supports local commits even offline.

**Remote repository:** Is hosted elsewhere, such as GitHub. You can send commits to it with `git push` and receive updates with `git fetch` or `git pull`.

A local commit is not automatically uploaded to the remote.

---

## 7. How Git and GitHub Work Together

1. Edit project files locally.
2. Inspect and stage changes with Git.
3. Commit changes to local history.
4. Push commits to a remote GitHub repository.
5. View files and commit history on GitHub.

```text
Local files -> git add -> git commit -> local Git history
                                      |
                                   git push
                                      v
                               GitHub remote

git fetch = download remote updates without integrating
git pull  = fetch and integrate remote updates
```

---

## 8. Important Git Commands

| Command | Purpose |
|---|---|
| `git status` | Show working-tree and staging state |
| `git add <file>` | Stage selected changes |
| `git commit -m "message"` | Record staged changes locally |
| `git log` | Display commit history |
| `git remote -v` | Show configured remote URLs |
| `git fetch` | Download remote updates without integrating |
| `git pull` | Fetch and integrate remote updates |
| `git push` | Send local commits to a remote |

These are Git commands. They interact with GitHub when the configured remote points to GitHub.

---

## 9. Understanding the origin Remote

When a repository is cloned, Git commonly names its default remote `origin`.

Run `git remote -v` to inspect it. Example output (illustrative):

```text
origin  https://github.com/username/project.git (fetch)
origin  https://github.com/username/project.git (push)
```

`origin` is a local nickname for the remote. The URL identifies the remote location. Fetch and push indicate the direction of data transfer.

---

## 10. Basic Local-to-Remote Workflow

```powershell
git status
git add "path/to/file"
git status
git commit -m "Describe the change"
git push origin main
```

1. Inspect repository state.
2. Stage the intended file.
3. Verify staged changes.
4. Create a local commit.
5. Push the commit to the remote `main` branch.

**Important:** `git commit` records locally; `git push` transfers commits to the remote.

## 13. Troubleshooting

**`git` is not recognized:** Check whether Git is installed, reopen the terminal after PATH changes, and verify with `git --version`.

**`git remote -v` shows no output:** The repository may not have a remote configured. Confirm the correct repository URL before adding one.

**Local and GitHub histories differ:** Some local commits may not be pushed, or the remote may contain commits not present locally. Inspect status and history before changing anything.

**Push is rejected:** Check authentication, write permission, branch state, and the full error. Avoid force-pushing unless you understand its impact and have a specific reason.

**A commit is visible locally but not on GitHub:** A commit remains local until pushed. Check the current branch and remote.
