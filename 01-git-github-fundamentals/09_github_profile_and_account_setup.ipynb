**Module:** 01 – Git & GitHub Fundamentals

## 1. Topic Introduction

A GitHub account lets you sign in and use GitHub. Your GitHub profile is your public developer identity. It can showcase your projects, skills, contributions, and professional interests.

## 2. Learning Objectives

- Explain the difference between a GitHub account and profile.
- Identify the main parts of a GitHub profile.
- Edit basic profile information.
- Explain profile README, pinned repositories, and the contribution graph.
- Identify basic account-security practices.

## 3. GitHub Account vs GitHub Profile

- **Account:** Your sign-in identity, authentication, and account settings.
- **Profile:** The page representing you on GitHub, showing information and work you choose to share.

A profile URL normally follows `https://github.com/username`.

## 4. GitHub Username and Profile URL

Your username identifies your GitHub account and appears in your profile URL. Choose a professional, readable username. Changing an established username can affect links and integrations, so consider the consequences first.

Example: username `octocat`; profile URL `https://github.com/octocat`.

## 5. Profile Picture and Bio

- **Profile picture:** An image associated with your account.
- **Bio:** A short description of your role, interests, or skills.

Example bio: `B.Tech CSE Student | Python | Data Science | AI/ML`

Keep your bio concise and accurate. Only share information you are comfortable making public.

## 6. Other Profile Information

Depending on what you choose to share, your profile can include your name, pronouns, location, website, and social or professional links. These fields are optional. Review what is public before adding personal information.

## 7. Pinned Repositories

Pinned repositories are selected repositories highlighted on your profile. They help visitors find projects you want to showcase.

Good choices include completed projects, useful learning repositories, and work that demonstrates your skills. Keep project names, descriptions, and README files clear.

## 8. Contribution Graph

The contribution graph summarizes eligible GitHub activity over time. It can include commits and other supported activities, subject to GitHub's contribution rules and repository settings.

Not every action appears in the graph, and a quiet graph does not by itself mean your work is missing.

## 9. Profile README

A profile README is a special README displayed on your GitHub profile.

To create one:
1. Create a repository named exactly the same as your GitHub username.
2. Make the repository public if you want the README publicly visible.
3. Add a file named `README.md`.
4. Add your introduction, skills, projects, and other information.

GitHub recognizes the matching repository and displays its README on your profile.

## 10. Editing Your Profile and Account Settings

To edit basic profile information:
1. Sign in to GitHub.
2. Open your profile menu and select **Your profile**.
3. Select **Edit profile**.
4. Update the fields you want to share and save.

Useful pages:
- Profile settings: https://github.com/settings/profile
- Security settings: https://github.com/settings/security

The exact interface may change as GitHub updates its website.

## 13. Troubleshooting

**Profile URL does not open**
- Check the spelling of the username and URL.
- Confirm that you are connected to the internet.

**Profile README is not displayed**
- Confirm that the repository name exactly matches your username.
- Confirm that it contains a file named `README.md`.
- Check repository visibility and GitHub's profile README requirements.

**Pinned repository is not visible**
- Check the profile's pinned section and whether you selected the repository to pin.

**Cannot change a setting**
- Confirm that you are signed in to the correct account and have access to that setting.

## 17. Sticky Notes

- Account = sign-in identity and account settings.
- Profile = public developer identity.
- Profile README repository name must exactly match your username.
- Pinned repositories showcase selected work.
- Contribution graph shows eligible activity, not every action.
- Use strong, unique passwords and enable 2FA.
- Never publish passwords, tokens, API keys, or recovery codes.
