## 03. What Is Git?

**Git** is a free, open-source, distributed version control system (VCS). It tracks changes to files and helps developers manage a project's history.

Git helps developers:

- Track changes over time.
- Review and compare versions.
- Restore earlier versions when needed.
- Work on separate features using branches.
- Collaborate with other developers.

**Example:** If a code change breaks a project, Git's history can help you inspect earlier commits and recover a working version.

**Remember:** Git is the tool that manages version history. GitHub is a separate online platform that can host Git repositories.

## 04. Why Was Git Created?

Git was created by **Linus Torvalds in 2005** to support development of the Linux kernel.

The project needed a version-control system that could support:

- Distributed development.
- Speed when working with a large project.
- Collaboration across a large developer community.

Git was designed to manage project history while allowing developers to work with their own local repositories.

## 05. Problems Git Solves

| Problem | How Git helps |
|---|---|
| Losing track of earlier code | Keeps a history of committed changes |
| Accidental or unwanted changes | Helps inspect and restore earlier versions |
| Multiple people changing a project | Supports collaboration and merging |
| Developing features separately | Provides branches |
| Understanding what changed | Provides history and comparison commands |
| Recording project milestones | Uses commits with messages and author information |

Git does not prevent every mistake. It provides tools and history that help developers manage and review changes.

## 06. How Git Works: Three Main Areas

### 6.1 Working Directory

The project files and folders you see and edit. New edits first appear in the Working Directory.

### 6.2 Staging Area (Index)

A preparation area where you select the changes that should be included in the next commit.

```bash
git add filename
```

### 6.3 Local Repository

The Git database that stores committed project snapshots and history.

```bash
git commit -m "Describe the change"
```

**Important:** These are the three main local areas in the basic Git workflow. A remote repository is hosted elsewhere; GitHub is one service that can host it.

## 07. The Basic Git Workflow

A simplified workflow:

```text
Working Directory
       |
       | git add
       v
  Staging Area
       |
       | git commit
       v
  Local Repository
       |
       | git push
       v
  Remote Repository (for example, GitHub)
```

1. Edit files in the Working Directory.
2. Select changes for the next commit using `git add`.
3. Record the staged changes in local history using `git commit`.
4. Send local commits to a remote repository using `git push`.

**Remember:** `git commit` records changes locally. It does not automatically upload them to GitHub.

## 08. Common Git Commands

| Command | Purpose |
|---|---|
| `git --version` | Display the installed Git version |
| `git init` | Initialize a Git repository |
| `git clone <url>` | Clone a remote repository |
| `git status` | Show the current repository state |
| `git add <file>` | Stage a file |
| `git add -A` | Stage additions, modifications, and deletions |
| `git commit -m "message"` | Create a commit |
| `git log` | View commit history |
| `git diff` | Compare unstaged changes |
| `git push` | Send local commits to a remote |
| `git pull` | Fetch and integrate remote changes |

**Note:** `git add`, `git commit`, `git push`, and `git pull` perform different steps. Staging, committing, and pushing are not interchangeable.

## 09. Important Git Terminology

| Term | Meaning |
|---|---|
| Git | Distributed version control system |
| VCS | Version Control System |
| Repository | Project location managed by Git, including its history |
| Working Directory | Checked-out project files you currently edit |
| Staging Area / Index | Area where selected changes are prepared for a commit |
| Commit | A recorded snapshot of staged changes |
| Branch | An independent line of development |
| Local Repository | Git repository stored on your computer |
| Remote Repository | Repository hosted elsewhere |
| Push | Send local commits to a remote repository |
| Pull | Fetch and integrate remote changes |
| Merge | Combine changes from different lines of development |

## 10. Understanding the Git Workflow with an Example

Imagine a project containing:

```text
MyProject/
├── main.py
├── calculator.py
└── README.md
```

After editing `main.py`, a typical workflow is:

| Step | Command | Purpose |
|---|---|---|
| 1 | `git status` | Inspect the current state |
| 2 | `git add main.py` | Stage the selected file |
| 3 | `git status` | Verify what is staged |
| 4 | `git commit -m "Update main.py"` | Save a commit locally |
| 5 | `git push origin main` | Send local commits to the remote `main` branch |

A commit records staged changes in the local repository. A push sends commits to the configured remote repository.

## 13. Troubleshooting

| Situation | What to check |
|---|---|
| `git` is not recognized | Check whether Git is installed and available in the system PATH. Reopen the terminal after installation or PATH changes. |
| `git status` says this is not a Git repository | Confirm that the terminal is inside the intended repository folder. |
| `git add` does not stage the expected file | Check the filename and current folder; then review `git status`. |
| `git commit` says there is nothing to commit | Check whether changes were made and staged. |
| `git push` is rejected | The remote may contain commits you do not have locally, or authentication/permissions may need attention. Read the full error before choosing a fix. |

Do not run reset, force-push, or cleanup commands as a first response to an error. Inspect the repository state first.
