## 1. Topic Introduction

GitHub is a cloud-based platform for hosting Git repositories. It helps developers store code online, collaborate with others, review changes, track work, and automate development workflows.

**Simple example:** Git tracks changes to your project on your computer. GitHub can host that Git repository online so you can access and share it.

## 3. What Is GitHub?

GitHub is a web-based platform that hosts Git repositories. A repository can contain source code, notebooks, documentation, images, configuration files, and other project resources.

GitHub adds collaboration and project-management features around Git, including pull requests, issues, discussions, and automation.

**Key point:** GitHub does not replace Git. It provides an online platform for working with Git repositories.

## 4. Why Developers Use GitHub

- **Backup and access:** Keep a remote copy of project work and access it from different devices.
- **Collaboration:** Work with other developers on shared projects.
- **Code review:** Discuss and review proposed changes before merging them.
- **Project tracking:** Record bugs, tasks, and feature requests with Issues.
- **Portfolio:** Share selected projects and learning work.
- **Automation:** Run workflows such as tests and deployment with GitHub Actions.
- **Open source:** Discover projects and contribute to them.

## 5. Git vs GitHub

| Git | GitHub |
|---|---|
| A distributed version control system | A platform for hosting Git repositories |
| Tracks changes to files | Stores remote repositories and supports collaboration |
| Commonly used through local commands | Used through a website, app, and Git integrations |
| Can work without an internet connection | Online features require network access |

**Remember:** Git is the version control tool; GitHub is one platform that hosts Git repositories.

## 6. Main GitHub Features

1. **Repositories:** Store project files and Git history.
2. **Pull Requests:** Propose changes and request review before merging.
3. **Issues:** Track bugs, tasks, questions, and improvements.
4. **GitHub Actions:** Automate workflows such as testing and deployment.
5. **Discussions:** Provide a place for community conversations when enabled.
6. **Projects:** Organize and track work using boards, tables, and roadmaps.
7. **Releases:** Publish versioned packages or project milestones.

## 7. Understanding a GitHub Repository

A repository is a project space that contains files and their version history.

Example structure:

```text
GitHub-learning/
├── README.md
├── 01-git-github-fundamentals/
│   ├── README.md
│   ├── 01_introduction_to_version_control.ipynb
│   ├── 02_what_is_git.ipynb
│   └── 03_what_is_github.ipynb
└── .gitignore
```

- `README.md`: Describes the project or folder.
- `.gitignore`: Lists files and folders Git should ignore.
- `.ipynb`: A Jupyter Notebook containing Markdown and code cells.

A repository can contain many folders and file types. The structure depends on the project's needs.

## 8. Public vs Private Repositories

| Public | Private |
|---|---|
| Anyone can view the repository | Only the owner and people granted access can view it |
| Useful for portfolios and open-source projects | Useful when work should have restricted access |
| Others can generally clone it | Access is controlled by repository permissions |

**Important:** Public visibility does not mean everyone can directly push changes to your repository. Write access still depends on permissions.

## 9. Important GitHub Terminology

| Term | Meaning |
|---|---|
| Repository (repo) | Project space containing files and Git history |
| Remote | A named reference to a repository hosted elsewhere |
| Clone | Create a local copy of a remote repository |
| Fork | Create your own GitHub-hosted copy of another repository |
| Branch | An independent line of development |
| Commit | A recorded set of changes in Git |
| Push | Send local commits to a remote repository |
| Pull | Fetch and integrate changes from a remote repository |
| Pull Request (PR) | Propose changes for review and merging |
| Issue | Track a bug, task, question, or improvement |
| Star | Bookmark or show appreciation for a repository |
| Watch | Subscribe to repository notifications |

## 10. Basic GitHub Collaboration Workflow

A common contribution workflow is:

1. Find the repository you want to contribute to.
2. Fork it if you do not have write access and the project accepts fork-based contributions.
3. Clone the repository to your computer.
4. Create a branch for your work.
5. Make changes and commit them with Git.
6. Push the branch to GitHub.
7. Open a Pull Request.
8. Discuss and revise the changes if needed.
9. The project maintainers review and may merge the Pull Request.

For a repository you own, you may work directly with its branches without forking.

## 13. Troubleshooting

**Problem: GitHub repository page shows 404.**
- Check that you are signed in to the correct account.
- Verify the repository owner and name.
- A private repository may not be visible to an account without access.

**Problem: Notebook does not render as expected.**
- Confirm that the file is a valid `.ipynb` notebook.
- Open it locally in Jupyter or VS Code to check its structure.
- Review the file's raw content if GitHub cannot render it.

**Problem: You cannot push changes.**
- Confirm that you have write permission.
- Check that you are pushing to the intended remote and branch.
- Review the authentication and Git error message before changing configuration.

**Problem: You cannot see a Pull Request or Issues tab.**
- Check repository settings and available features. Some tabs can be disabled or hidden.
