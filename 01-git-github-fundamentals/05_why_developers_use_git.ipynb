# Topic 05 – Why Developers Use Git

## 1. Topic Introduction

Git is a version control system that helps developers record changes to project files. Developers use it to manage code history, experiment safely, collaborate with others, and recover earlier versions when needed.

In this notebook, you will learn the practical reasons developers use Git and practise a few commands in your existing Git practice repository.

## 3. Why Developers Need Version Control

Software projects change continuously. Developers add features, fix bugs, update documentation, and sometimes need to undo changes.

Without version control, developers may create many manually named copies such as `project-final`, `project-final-new`, and `project-final-fixed`. This becomes difficult to manage.

Version control records changes in an organised way so developers can understand how a project evolved and return to earlier saved states.

## 4. Benefit 1 – Track Changes and Project History

Git records project snapshots as commits. A commit includes a message and identifies the saved state of the tracked project files.

The history helps answer questions such as:

- What changed?
- When was it committed?
- Which commit introduced a particular change?
- What was the project state before the change?

Useful command:

```powershell
git log --oneline
```

This displays a compact list of commits.

## 5. Benefit 2 – Recover from Mistakes

A developer may accidentally introduce a bug or make an unwanted change. Git provides ways to inspect earlier commits and restore files or create a new commit that reverses a previous change.

Important distinction:

- `git diff` helps inspect changes that have not been committed.
- `git restore` can discard selected uncommitted working-tree changes.
- `git revert` creates a new commit that reverses the effect of an earlier commit.
- `git reset` moves a branch reference and can affect commit history; use it carefully.

Do not run destructive commands on important work unless you understand their effects and have a backup.

## 6. Benefit 3 – Work on Features with Branches

A branch is an independent line of development. Developers can create a branch to work on a feature or bug fix without making those changes directly on the main development branch.

Example:

```powershell
git branch
git switch -c feature-login
```

The second command creates and switches to a branch named `feature-login`.

When the work is ready, it can be reviewed and merged into another branch.

## 7. Benefit 4 – Collaborate with Other Developers

Git helps multiple developers work on a shared project. Each developer can make commits locally and share changes through a remote repository.

A typical collaboration process includes:

1. Get the project from a remote repository.
2. Create a branch for a task.
3. Make and commit changes.
4. Push the branch to the remote repository.
5. Open a pull request for review.
6. Merge approved changes.

Git tracks changes; hosting platforms such as GitHub provide tools for pull requests, code review, issue tracking, and project collaboration.

## 8. Benefit 5 – Experiment Safely

Branches allow developers to try new ideas separately. If an experiment does not work, they can switch back to another branch without mixing unfinished work into the main branch.

Before switching branches, check your working tree:

```powershell
git status
```

Commit or safely handle important changes before switching. Git may prevent a switch if uncommitted changes would be overwritten.

## 9. Benefit 6 – Improve Accountability and Team Communication

Commit messages provide a readable explanation of why a change was made. A clear history helps teammates review work and understand decisions.

Good commit message:

```text
Add input validation to registration form
```

Less useful commit message:

```text
changes
```

Write short, specific messages that describe the purpose of the change.

## 10. Basic Git Workflow

A common local workflow is:

```powershell
git status
git add filename
git commit -m "Describe the change"
git log --oneline
```

Meaning:

1. `git status` checks the working tree.
2. `git add` stages the selected file.
3. `git commit` records the staged changes.
4. `git log` displays commit history.

To share commits with a remote repository, a developer may also use:

```powershell
git push
```

The exact push command depends on the repository's configured remote and branch.

## 13. Troubleshooting

**Problem: `fatal: not a git repository`**

Cause: The terminal is not inside a Git repository.

Solution: Change directory to your existing Git practice repository, then retry:

```powershell
cd "C:\Users\Nikhil\01_Learning\02_Git_&_GitHub\01_Git_Practice"
```

If your local folder has moved, use its current path instead.

**Problem: `git log` shows a pager or does not return to the prompt**

Press `q` to exit the Git log viewer.

**Problem: `git diff` shows no output**

There may be no unstaged changes to display. Check the state with:

```powershell
git status
```

**Problem: Cannot switch branches because local changes would be overwritten**

Inspect the changes with `git status` and `git diff`. Commit, stash, or otherwise safely preserve the work before switching. Do not discard changes unless you are certain they are no longer needed.
