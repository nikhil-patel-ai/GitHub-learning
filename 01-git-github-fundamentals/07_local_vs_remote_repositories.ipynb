# Topic 07 – Local vs Remote Repositories

## 1. Topic Introduction

Git repositories can be stored on your computer or hosted on another system. These are commonly called local and remote repositories.

Understanding the difference helps you work offline, maintain project history, share commits, and synchronize work with GitHub.

## 2. Learning Objectives

After completing this topic, you should be able to:

- Explain what local and remote repositories are.
- Compare their locations and purposes.
- Explain how local and remote repositories are synchronized.
- Use Git commands to inspect remotes and exchange changes.
- Understand common synchronization errors at a basic level.

## 3. What Is a Local Repository?

A local repository is a Git repository stored on your computer. You use it to edit files, stage changes, create commits, inspect history, and work with branches.

Example local project path:

```text
C:\Users\Nikhil\NVSCode\02_Git_&_GitHub
```

Local Git operations such as `git status`, `git add`, `git commit`, and `git log` generally do not require an internet connection.

## 4. What Is a Remote Repository?

A remote repository is another copy of a Git repository hosted on a server or service. It allows you to share work and synchronize commits across computers or with other developers.

Example remote repository:

```text
https://github.com/nikhil-patel-ai/GitHub-learning
```

GitHub is a hosting and collaboration platform. Git is the version-control system used to manage the repository history.

## 5. Local vs Remote Repositories

| Feature | Local repository | Remote repository |
|---|---|---|
| Location | Your computer | A server or hosting service |
| Main purpose | Local development and history | Sharing and synchronization |
| Commit | Create commits locally | Receives commits through push |
| Internet | Not needed for most local operations | Needed to communicate with it |
| Collaboration | Can work independently | Can share changes with others |

A local repository and a remote repository are separate copies. Changes in one do not automatically appear in the other.

## 6. How Are They Connected?

Git stores a named reference to a remote repository. The common default name is `origin`.

Inspect configured remotes:

```powershell
git remote -v
```

Typical output format:

```text
origin  https://github.com/OWNER/REPOSITORY.git (fetch)
origin  https://github.com/OWNER/REPOSITORY.git (push)
```

The URL above is an example format. Your actual output depends on your repository configuration.

A remote is a convenient name and URL for another repository. It is not a separate branch.

## 7. How Synchronization Works

A common workflow is:

1. Edit files in the local working directory.
2. Stage the changes.
3. Commit the changes to local repository history.
4. Push local commits to the remote repository.
5. Fetch or pull changes that other contributors have pushed.

Remember: a commit saves a change in local Git history. It does not automatically upload that commit to GitHub.

## 8. Push, Fetch, and Pull

**Push**

```powershell
git push origin main
```

Sends local commits from the current local `main` branch to the remote branch named `main` on `origin`.

**Fetch**

```powershell
git fetch origin
```

Downloads updates and remote-tracking information from `origin`. It does not integrate those changes into the current branch.

**Pull**

```powershell
git pull origin main
```

Fetches updates and integrates them into the current local branch. Depending on configuration and situation, Git may use a merge or rebase strategy.

## 9. Important Terminology

| Term | Meaning |
|---|---|
| Local repository | Git repository on your computer |
| Remote repository | Git repository hosted elsewhere |
| Remote | A named reference to another repository |
| `origin` | Common default remote name |
| Push | Send local commits to a remote |
| Fetch | Download remote updates without integrating them |
| Pull | Fetch and integrate remote updates |
| Remote-tracking branch | Local reference that records the state of a branch on a remote |
| Synchronization | Bringing local and remote repository histories up to date |

## 10. Basic Local-to-Remote Workflow

For a file you have changed:

```powershell
git status
git add filename
git commit -m "Describe the change"
git push origin main
```

Replace `filename` with the actual relative file path. This workflow assumes the remote and branch are configured and that the remote can accept the push.

To inspect remote changes before integrating them:

```powershell
git fetch origin
git status
```

If the remote contains commits that your local branch does not have, integrate them using the appropriate pull or rebase workflow before pushing.

## 13. Troubleshooting

**Problem: `fatal: not a git repository`**

Cause: The terminal is not inside a Git repository.

Solution: Change directory to your actual local repository folder and retry the command.

**Problem: `error: remote origin already exists`**

Cause: A remote named `origin` is already configured.

Solution: Inspect it with:

```powershell
git remote -v
```

Do not add it again. If its URL is incorrect and you intentionally need to change it, use `git remote set-url origin <correct-url>`.

**Problem: Push rejected with `non-fast-forward`**

Cause: The remote branch contains commits that your local branch does not have.

Solution: Fetch and inspect the remote changes, then integrate them with the appropriate merge or rebase workflow before pushing. Do not force-push as a routine fix.

**Problem: `git fetch origin` cannot connect**

Cause: Network connectivity, authentication, permissions, or the remote URL may be incorrect.

Solution: Check your internet connection and inspect the configured URL using `git remote -v`. Verify repository access and authentication.

**Problem: `git branch -r` shows no remote branches**

Cause: Remote-tracking references may not have been fetched yet, or the remote may not have branches available to your account.

Solution: Verify the remote with `git remote -v`, then run `git fetch origin` and inspect again.
