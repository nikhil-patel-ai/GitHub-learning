# 02. Verifying Git Installation

## 1. Learning Objectives
After completing this topic, you will be able to:
- Check whether Git is installed and accessible.
- Identify the installed Git version.
- Locate the Git executable on Windows.
- Verify Git access from PowerShell and Git Bash.
- Recognize common installation-verification errors.

## 2. Why Verify Git?
After installing Git, verify that Windows can find and run it. This confirms that the command-line tools are accessible before you start using Git in projects.

Since Git commands have already worked in your PowerShell session, this topic verifies the installation rather than asking you to reinstall Git.

## 3. Verify Git Version

**Command:**
```powershell
git --version
```

**Purpose:** Displays the installed Git version.

**Example expected output** (your version may be different):
```text
git version 2.x.x.windows.1
```

Record your actual output in the practice record below. Do not copy the example as your actual result.

## 4. Locate Git Executable

**Command:**
```powershell
where.exe git
```

**Purpose:** Lists the locations of `git.exe` that Windows finds through PATH.

**Example expected output:**
```text
C:\Program Files\Git\cmd\git.exe
```

Your installation path may differ.

You can also use PowerShell's command discovery:

```powershell
Get-Command git
```

This shows the command PowerShell resolves when you type `git`.

## 5. Inspect Git Executable Path

**Command:**
```powershell
git --exec-path
```

**Purpose:** Displays the path to Git's internal helper programs. This is not the same as the location of `git.exe`.

## 6. Verify Git in PowerShell and Git Bash
1. Open PowerShell and run `git --version`.
2. Open Git Bash from the Windows Start Menu.
3. Run `git --version` in Git Bash too.
4. Compare the version strings.

If both terminals report a Git version, Git is accessible from both environments.

## 7. Practical Activity – Your Windows Laptop

Run these commands one at a time in PowerShell:

```powershell
git --version
where.exe git
Get-Command git
git --exec-path
```

Then open Git Bash and run:

```bash
git --version
```

## 8. Understanding the Results

| Command | What it verifies |
|---|---|
| `git --version` | Git runs and reports a version |
| `where.exe git` | Windows can locate `git.exe` through PATH |
| `Get-Command git` | PowerShell can resolve the Git command |
| `git --exec-path` | Git can locate its internal helper programs |

## 9. Common Troubleshooting

### `git` is not recognized
- Close and reopen PowerShell.
- Run `where.exe git`.
- If no path is returned, inspect whether Git for Windows is installed and whether its command-line directory is included in PATH.

### `where.exe git` returns more than one path
Multiple Git installations may be available. Use `Get-Command git` to see which command PowerShell resolves. Do not remove installations until you understand which applications depend on them.

### PowerShell and Git Bash show different versions
There may be multiple Git installations or different PATH configurations. Compare the executable locations and investigate before changing PATH.

### Git Bash is not available
Check whether Git for Windows was installed with Git Bash components. Do not reinstall if Git Bash is already available.
