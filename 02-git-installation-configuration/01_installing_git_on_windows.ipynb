# 01. Installing Git on Windows

## 1. Learning Objectives
After completing this topic, you will be able to:
- Explain why Git must be installed before using Git commands.
- Download Git from its official website.
- Understand the main choices in the Windows installer.
- Install Git for command-line use and editor integration.

## 2. What is Git Installation?
Git is a version control system. Installing Git adds the Git program to Windows so you can run commands such as `git status`, `git add`, `git commit`, and `git push`.

**Official download:** https://git-scm.com/downloads/win

Download Git only from the official Git website.

## 3. Before Installation
- Use an account that has permission to install software.
- Close terminals that are currently open; reopen them after installation so they can detect the updated PATH.
- If Git is already installed and working, do not reinstall it just for this lesson. Study the installation process and continue with verification in Topic 02.

## 4. Installing Git on Windows
1. Open the official download page: https://git-scm.com/downloads/win
2. Download the installer for your Windows system.
3. Run the downloaded installer.
4. Review the license and continue.
5. Keep the default destination unless you have a specific reason to change it.
6. Review the component and Start Menu options.
7. When asked about the default editor, choose an editor you actually use. Visual Studio Code is a suitable choice if it is installed and available in the list.
8. For PATH configuration, choose the option that allows Git from the command line and also from third-party software. This makes Git usable in terminals and development tools.
9. For the SSH executable and HTTPS transport backend, the recommended defaults are generally suitable.
10. Keep the default line-ending and terminal choices unless your organization or project requires another configuration.
11. Review the remaining options, install, and finish.

> Installer screens and option wording can change between Git releases. Read each screen before selecting an option; do not blindly click through it.

## 5. Important Installation Choices

| Setting | Purpose |
|---|---|
| Default editor | Editor opened by Git for commit messages and other text |
| PATH environment | Makes `git` available from terminals and other applications |
| SSH executable | Supports SSH-based communication with remote repositories |
| HTTPS backend | Supports HTTPS-based communication with remote repositories |
| Line endings | Controls handling of Windows and Unix-style line endings |
| Terminal emulator | Determines the terminal experience used by Git Bash |

## 6. Git Bash and Command-Line Access
Git for Windows includes Git Bash, a terminal that provides a Bash-like environment on Windows. Git can also be used from PowerShell, Command Prompt, and supported development tools when PATH is configured appropriately.

## 9. Common Troubleshooting

### `git` is not recognized
- Close and reopen the terminal.
- Check whether Git for Windows is installed.
- Check that the Git command-line option was included in PATH.
- If necessary, rerun the installer and review PATH configuration.

### Git works in Git Bash but not PowerShell
- Reopen PowerShell.
- Check whether Git's command-line path is available in the Windows PATH.
- Verify the installation configuration before changing environment variables manually.

### Installer asks for administrator permission
- This may be expected when installing for all users or into a protected location.
- Proceed only if you have permission to install software on that computer.
