## 2. Introduction

To work with Git, we need a command-line interface (CLI) or a graphical user interface (GUI).

A **CLI** lets us interact with software by typing commands. Git provides commands such as `git status`, `git add`, and `git commit`.

Git commands are not limited to Git Bash. They can also be run from other terminals, provided Git is installed and available in the system's PATH.

## 3. What is Git Bash?

**Git Bash** is an application installed with Git for Windows. It provides a Bash command-line environment on Windows, along with common Unix-style commands.

Examples of commands commonly used in Git Bash:

- `pwd` — show the current directory.
- `ls` — list files and folders.
- `cd` — change directory.
- `mkdir` — create a directory.
- `clear` — clear the terminal screen.

Git Bash is useful when following tutorials that use Linux/macOS-style terminal commands.

## 4. What is Git CLI?

**Git CLI (Command-Line Interface)** means using Git by typing Git commands in a terminal.

Examples:

```bash
git --version
git status
git add .
git commit -m "Add project files"
git log
```

Git CLI is the way we interact with Git through commands. Git Bash is one environment where we can run those commands.

## 5. Git Bash vs Git CLI

| Git Bash | Git CLI |
|---|---|
| A terminal environment provided by Git for Windows. | The command-based way of using Git. |
| Provides Bash and Unix-style commands. | Provides Git commands such as `git status`. |
| Windows-focused application. | Can be used in multiple terminals and operating systems. |
| Example: `ls`, `pwd`, `git status`. | Example: `git status`, `git add`, `git commit`. |

**Remember:** Git Bash is a place where you can run Git CLI commands. They are related, but they are not the same thing.

## 6. Terminal, Shell, and CLI

These terms are related:

- **Terminal:** The application or interface where you type commands.
- **Shell:** The program that interprets commands typed into the terminal. Bash and PowerShell are examples.
- **CLI:** A way of interacting with software through text commands. Git CLI is Git's command-based interface.

For example, Windows Terminal can host PowerShell, and Git Bash can provide a Bash shell.

## 7. Where Can We Run Git Commands?

If Git is installed correctly and available in PATH, Git commands can run in:

1. Git Bash
2. Windows PowerShell
3. Command Prompt (CMD)
4. VS Code integrated terminal

The shell-specific commands may differ. For example, `ls` is commonly used in Git Bash, while `dir` is available in CMD and PowerShell. Git commands such as `git status` work across these terminals.

## 8. Opening Git Bash on Windows

You can open Git Bash in either of these ways:

1. Search for **Git Bash** in the Windows Start menu.
2. In File Explorer, open a project folder, right-click an empty area, and choose **Open Git Bash here** (wording can vary by version).

To check that Git is available, run:

```bash

```

Expected output format (the version number may differ):


This is an example of expected output, not a claim about your installed version.

## 9. Basic Git Commands – Practical Reference

Run these commands from inside a Git repository where appropriate.

| Command | Purpose |
|---|---|
| `git --version` | Display the installed Git version. |
| `git status` | Show the current repository status. |
| `git init` | Initialize a new Git repository. |
| `git add <file>` | Stage a file. |
| `git add .` | Stage changes in the current directory and its subdirectories. |
| `git commit -m "message"` | Create a commit from staged changes. |
| `git log --oneline` | Display a compact commit history. |
| `git branch` | List local branches. |
| `git remote -v` | Display configured remotes. |

**Important:** Do not run `git init` in an existing repository unless you intentionally want to initialize or reinitialize repository metadata. For this lesson, use `git status` and other inspection commands in your existing learning repository.

## 10. Git Bash vs PowerShell vs CMD

| Feature | Git Bash | PowerShell | CMD |
|---|---|---|---|
| Shell | Bash | PowerShell | Windows command shell |
| Common list command | `ls` | `Get-ChildItem` / `ls` alias | `dir` |
| Current directory | `pwd` | `Get-Location` / `pwd` alias | `cd` |
| Git commands | Supported | Supported | Supported |
| Typical use | Git and Unix-style commands | Windows administration and scripting | Traditional Windows commands |

The exact aliases available can depend on the shell and system configuration.

## 11. Common Errors and Troubleshooting

### Error: `git is not recognized` or `command not found`

**Possible cause:** Git is not installed, or its executable directory is not available through PATH.

**What to check:**

1. Run `git --version`.
2. If it fails, confirm Git for Windows is installed.
3. Close and reopen the terminal after installation or PATH changes.
4. If it still fails, inspect the Git installation and PATH configuration.

### Error: `fatal: not a git repository`

**Cause:** The current directory is not inside a Git repository.

**Solution:** Use `cd` to move into the correct project folder, then run `git status`.

### Error: A terminal command works in Git Bash but not in CMD

**Cause:** The command may be specific to Bash or another shell.

**Solution:** Use the equivalent command for the current shell, or run it in Git Bash.
