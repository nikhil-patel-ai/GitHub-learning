# 04. Configuring Username and Email

## 1. Learning Objectives
By the end of this topic, you will be able to:
- Explain why Git stores a username and email with commits.
- Configure and inspect global and repository-local identity.
- Understand why local identity overrides global identity.
- Identify the configuration file that supplies each value.
- Remove a temporary local identity and return to the global identity.

## 2. Key Concepts
- **`user.name`**: Author name recorded in Git commits.
- **`user.email`**: Author email recorded in Git commits.
- **Global scope**: Applies to repositories for the current Windows user unless overridden.
- **Local scope**: Applies only to the current repository and overrides the global value.
- **Effective identity**: The value Git actually uses in the current repository.

Git identity is commit metadata; it is not the same as your GitHub sign-in password or authentication token.

## 3. Configuration Scopes and Precedence

| Scope | Configuration file | Applies to |
|---|---|---|
| System | Git installation's system config | All users on the computer |
| Global | `C:\Users\<user>\.gitconfig` | Current Windows user |
| Local | `<repository>\.git\config` | One repository |

When the same key exists at multiple scopes, Git uses the most specific value:

**Local → Global → System**

For this practical, the global identity was already configured, so no global setting needed to be changed.

## 4. Practical A – Inspect Existing Identity

**Execution location:** PowerShell, inside  
`C:\Users\Nikhil\NVSCode\02_Git_&_GitHub`

Commands used:

```powershell
git config --global user.name
git config --global user.email
git config --local user.name
git config --local user.email
git config user.name
git config user.email
git --no-pager config --show-origin --list
```

### Actual Results

| Check | Actual result |
|---|---|
| Global username | `Nikhil Patel` |
| Global email | `knikhilpatel7281@gmail.com` |
| Local username before test | Not configured (blank output) |
| Local email before test | Not configured (blank output) |
| Effective username before test | `Nikhil Patel` |
| Effective email before test | `knikhilpatel7281@gmail.com` |

The global values came from `C:/Users/Nikhil/.gitconfig`. The local repository file initially contained no `user.name` or `user.email` entries.

## 5. Practical B – Test Repository-Local Identity

The following temporary values were used to demonstrate precedence:

```powershell
git config --local user.name "Local Test User"
git config --local user.name "Nikhil"
git config --local user.email "knikhilpatel7281@gmail.com"

git config --local user.name
git config --local user.email
git config user.name
git config user.email
git --no-pager config --show-origin --list
```

### Actual Results

| Setting | Global value | Local value during test | Effective value |
|---|---|---|---|
| `user.name` | `Nikhil Patel` | `Nikhil` | `Nikhil` |
| `user.email` | `knikhilpatel7281@gmail.com` | `knikhilpatel7281@gmail.com` | `knikhilpatel7281@gmail.com` |

**Observation:** The local username `Nikhil` overrode the global username `Nikhil Patel`. The local email was explicitly set to the same email as the global value, so the effective email looked unchanged.

The username was first set to `Local Test User`, then replaced with `Nikhil` before verification. The final local username during the test was therefore `Nikhil`.

## 6. Practical C – Restore Global Identity

After demonstrating local precedence, the temporary local settings were removed:

```powershell
git config --local --unset user.name
git config --local --unset user.email

git config --local user.name
git config --local user.email

git config user.name
git config user.email

git --no-pager config --show-origin --list
git status
```

### Actual Results

- Local username: blank (not configured).
- Local email: blank (not configured).
- Effective username: `Nikhil Patel`.
- Effective email: `knikhilpatel7281@gmail.com`.
- Global identity is again used by this repository.
- `git status` reported `On branch main`, branch up to date with `origin/main`, and `nothing to commit, working tree clean`.

**Conclusion:** Temporary local values were successfully removed. No commit was created during this identity test.

## 7. Important Commands – Quick Reference

| Command | Purpose |
|---|---|
| `git config --global user.name` | Read global username |
| `git config --global user.email` | Read global email |
| `git config --local user.name` | Read local username |
| `git config --local user.email` | Read local email |
| `git config user.name` | Read effective username |
| `git config user.email` | Read effective email |
| `git config --global user.name "Name"` | Set global username |
| `git config --global user.email "email"` | Set global email |
| `git config --local user.name "Name"` | Set repository-local username |
| `git config --local user.email "email"` | Set repository-local email |
| `git config --local --unset user.name` | Remove local username |
| `git config --local --unset user.email` | Remove local email |
| `git --no-pager config --show-origin --list` | Show settings and their source without opening a pager |

## 8. Troubleshooting

### `git config --local user.name` prints nothing
This means the repository has no local username configured. Git may inherit the global value.

### Effective identity differs from global identity
Check for a local override:

```powershell
git config --local user.name
git config --local user.email
```

### Output opens with `(END)`
Git launched a pager. Press `q` to exit, or use:

```powershell
git --no-pager config --show-origin --list
```

### `git config --local --unset` reports that a key is not found
That key is not configured locally. Check it with `git config --local --list` before attempting removal.

## 11. Summary

- Git records author name and email in commit metadata.
- Global identity applies by default to the current Windows user.
- Local identity applies to one repository and overrides global identity.
- `--show-origin` helps identify where a setting is stored.
- Removing local overrides restores the global identity when no other higher-priority setting applies.
