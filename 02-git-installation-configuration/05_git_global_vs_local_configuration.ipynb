## 1. Learning Objectives
- Understand Global and Local Git configuration.
- Identify where each configuration is stored.
- Understand configuration scope and precedence.
- Inspect configuration values and their source.

## 2. Key Terminology
- **System:** Settings applied to all users on the computer.
- **Global:** Settings applied to all repositories for the current Windows user.
- **Local:** Settings applied only to the current repository.
- **Effective value:** The value Git uses after applying configuration precedence.
- **Configuration origin:** The file from which a setting is loaded.

## 3. Concept Explanation

### A. Global Configuration
Global settings apply to all Git repositories for the current Windows user.

```powershell
git config --global user.name "Nikhil Patel"
git config --global user.email "knikhilpatel7281@gmail.com"
```

Global configuration is normally stored in:

`C:\Users\Nikhil\.gitconfig`

### B. Local Configuration
Local settings apply only to the current repository.

```powershell
git config --local user.name "Project User"
git config --local user.email "project@example.com"
```

Local configuration is stored in the repository's:

`.git/config`

### C. Configuration Precedence
When the same setting is defined at multiple levels, the more specific level takes precedence:

**Local > Global > System**

For this topic, the practical focus is Global and Local configuration.

## 4. Global vs Local Comparison

| Feature | Global | Local |
|---|---|---|
| Scope | All repositories for the current user | Current repository only |
| Option | `--global` | `--local` |
| Typical file | `~/.gitconfig` | `.git/config` |
| Priority | Lower than Local | Higher than Global |

## 5. Commands and Syntax

| Command | Purpose |
|---|---|
| `git config --global user.name` | Read global username |
| `git config --global user.email` | Read global email |
| `git config --local user.name` | Read local username |
| `git config --local user.email` | Read local email |
| `git config user.name` | Read effective username |
| `git config user.email` | Read effective email |
| `git --no-pager config --show-origin --list` | List settings and their source files |

Because Local username and email were not configured, Git inherited the Global values. If a Local value is set for this repository, it overrides the corresponding Global value only in this repository.

## 7. Common Mistakes and Troubleshooting
- **Mistake:** Using `--global` when you intended to configure only one project.  
  **Fix:** Use `--local` from inside the target repository.
- **Mistake:** Assuming a blank Local result means Git has no identity.  
  **Fix:** Check the effective value with `git config user.name` and `git config user.email`.
- **Mistake:** Looking for Global settings in `.git/config`.  
  **Fix:** Check the user's `.gitconfig` file.

## 8. Best Practices
- Use Global identity for your usual personal Git identity.
- Use Local identity only when a particular repository needs a different identity.
- Inspect the effective value before creating commits in an unfamiliar repository.
- Do not share private credentials or access tokens in configuration output.

## 10. Interview Questions and Answers

**Q1. What is Global Git configuration?**  
Global configuration stores settings that apply to all repositories for the current user.

**Q2. What is Local Git configuration?**  
Local configuration stores settings for one specific repository.

**Q3. Which has higher priority: Global or Local?**  
Local has higher priority than Global.

**Q4. Where is Global configuration stored on this Windows setup?**  
`C:\Users\Nikhil\.gitconfig`

**Q5. How can you identify the source of a Git setting?**  
Use `git config --show-origin --list`.

## 12. Summary
- Global configuration applies to all repositories for the current user.
- Local configuration applies only to the current repository.
- Local overrides Global when both define the same setting.
- `git config user.name` and `git config user.email` show the effective identity.
- `git config --show-origin --list` helps identify where settings come from.
