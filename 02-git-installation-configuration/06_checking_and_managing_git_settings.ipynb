# 06. Checking and Managing Git Settings

## 1. Learning Objectives
After completing this topic, you will be able to:
- Display Git configuration settings.
- Identify the source file for a setting.
- Retrieve effective and scope-specific values.
- Update or remove configuration entries safely.

## 2. Key Concepts
- **Effective setting:** The value Git uses in the current repository after applying configuration precedence.
- **Configuration source:** The file or scope where a setting is defined.
- **Get:** Reads a configuration value without changing it.
- **Unset:** Removes a configuration entry; it does not uninstall Git.

## 3. Important Commands

| Command | Purpose |
|---|---|
| `git config --list` | Display configuration settings |
| `git config --show-origin --list` | Display settings and their source files |
| `git config --get user.name` | Get the effective username |
| `git config --get user.email` | Get the effective email |
| `git config --global --get user.name` | Get the global username |
| `git config --local --get user.name` | Get the local username |
| `git config --get-regexp "^user\\."` | Display matching user-related settings |
| `git config --edit --global` | Open the global configuration file for editing |

## 6. Practical 2 – Understand Updating and Removing Settings

**Examples only – do not run unless you intentionally want to change your configuration.**

Set a global username:

```powershell
git config --global user.name "New Name"
```

Set a local email for the current repository:

```powershell
git config --local user.email "newemail@example.com"
```

Remove a global username entry:

```powershell
git config --global --unset user.name
```

Remove a local email entry:

```powershell
git config --local --unset user.email
```

**Important:** `--unset` removes the specified configuration entry. Before changing or removing a real setting, inspect its scope and source. Removing a global value may cause Git to use a lower-priority value, if one exists.

## 7. Troubleshooting

### A setting is not displayed
The setting may not be configured at the scope being queried. Check the effective value and inspect the source list.

### The effective value differs from the global value
A higher-priority configuration, such as a repository-local value, may override the global value.

### A command reports that a value is not found
That exact key may not exist at the requested scope. Inspect `git config --show-origin --list` before making changes.

- Use `--show-origin` to identify where settings come from.
- Use `--get` to retrieve a value.
- Use `--unset` to remove a specific entry.
- Inspect scope and source before changing configuration.
